## Computing Laplace Transforms with SymPy

### Example 1: Basic Laplace Transform Calculation

```python
import sympy as sp

# Define symbolic variables
t, s = sp.symbols('t s', positive=True)

# Define the function
f = t**2 * sp.exp(-3*t)

# Calculate the Laplace transform: returns (F, a, cond),
# where Re(s) > a is the region of convergence
F, a, cond = sp.laplace_transform(f, t, s)

print("Function:", f)
print("Laplace transform:", F)
print("Convergence abscissa:", a)

# Verify using the definition (integration)
F_manual = sp.integrate(sp.exp(-s*t) * f, (t, 0, sp.oo))
print("Verified transform:", sp.simplify(F_manual))
```

### Example 2: Transform of Trigonometric Functions

```python
import sympy as sp

# Define symbolic variables
t, s, omega = sp.symbols('t s omega', real=True, positive=True)

# Define the function
f = sp.sin(omega * t)

# Calculate the Laplace transform
F = sp.laplace_transform(f, t, s)

print("Function:", f)
print("Laplace transform:", F[0])
print("Convergence abscissa:", F[1])

# Simplify the result
F_simplified = sp.simplify(F[0])
print("Simplified transform:", F_simplified)
```

### Example 3: Using Laplace Transform Properties

```python
import sympy as sp

# Define symbols
t, s, a = sp.symbols('t s a', real=True)

# Define functions
f = t**2
g = sp.exp(-t)

# Calculate their Laplace transforms
F = sp.laplace_transform(f, t, s)[0]
G = sp.laplace_transform(g, t, s)[0]

# Verify linearity property
combined_function = 3*f + 2*g
combined_transform = sp.laplace_transform(combined_function, t, s)[0]
linearity_check = 3*F + 2*G

print("Combined function transform:", sp.simplify(combined_transform))
print("Using linearity property:", sp.simplify(linearity_check))
print("Match:", sp.simplify(combined_transform) == sp.simplify(linearity_check))

# Verify differentiation property
df_dt = sp.diff(f, t)
df_transform = sp.laplace_transform(df_dt, t, s)[0]
diff_property = s*F - f.subs(t, 0)

print("\nDerivative transform:", sp.simplify(df_transform))
print("Using differentiation property:", sp.simplify(diff_property))
print("Match:", sp.simplify(df_transform) == sp.simplify(diff_property))
```

## The Inverse Laplace Transform

### Using SymPy for Inverse Laplace Transforms

```python
import sympy as sp

# Define symbolic variables
s = sp.symbols('s', real=True)
t = sp.symbols('t', positive=True)

# Define the s-domain function
F_s = (3*s + 2)/(s**2 + 4*s + 13)

# Calculate the inverse Laplace transform
f_t = sp.inverse_laplace_transform(F_s, s, t)

print("S-domain function:", F_s)
print("Time domain function:", f_t)

# Verify by taking the Laplace transform of the result
F_s_check = sp.laplace_transform(f_t, t, s)[0]
print("Verification (should match original):", sp.simplify(F_s_check))
```

## The Convolution Theorem

### Example: Using the Convolution Theorem

```python
import sympy as sp

# Define variables
s, tau = sp.symbols('s tau', real=True)
t = sp.symbols('t', positive=True)

# Define the s-domain function
F_s = 1/(s*(s**2 + 1))

# Method 1: Direct inverse transform
f_t = sp.inverse_laplace_transform(F_s, s, t)
print("f(t) using direct inverse:", f_t)

# Method 2: Using convolution theorem
f1_t = 1  # inverse of 1/s
f2_t = sp.sin(t)  # inverse of 1/(s^2+1)

# Calculate convolution integral
convolution = sp.integrate(f1_t * f2_t.subs(t, t-tau), (tau, 0, t))
print("f(t) using convolution:", convolution)

# Verify they match
print("Match:", sp.simplify(f_t - convolution) == 0)
```

## Applications in Engineering

### Solving Differential Equations

#### Example: Mass-Spring-Damper System

```python
import sympy as sp

# Define symbols
t, s = sp.symbols('t s', positive=True)
m, c, k = 1, 4, 5  # System parameters
x0, v0 = 0, 0      # Initial conditions

# External force and its Laplace transform
F_t = 10 * sp.sin(2*t)
F_s = sp.laplace_transform(F_t, t, s)[0]

# Transfer function denominator
denominator = m*s**2 + c*s + k

# Solve for X(s)
X_s = (F_s + m*s**2*x0 + m*s*v0 + c*x0) / denominator

# Find x(t) by inverse Laplace transform
x_t = sp.inverse_laplace_transform(X_s, s, t)

print("X(s) =", X_s)
print("x(t) =", sp.simplify(x_t))

# Plot the solution
import numpy as np
import matplotlib.pyplot as plt

# Convert symbolic expression to numerical function
x_t_func = sp.lambdify(t, x_t, 'numpy')

# Create time points and evaluate
t_values = np.linspace(0, 10, 1000)
x_values = x_t_func(t_values)

plt.figure(figsize=(10, 6))
plt.plot(t_values, x_values)
plt.title('Mass-Spring-Damper System Response')
plt.xlabel('Time (s)')
plt.ylabel('Displacement x(t) (m)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Python Implementation for Plotting Step Functions

```python
import numpy as np
import matplotlib.pyplot as plt

# Create our own unit step function
def unit_step(x):
    return np.heaviside(x, 1)  # numpy's heaviside with value 1 at x=0

# Create time vector
t = np.linspace(-1, 10, 1000)

# Define the function using step functions
def f(t):
    # First, create the step functions
    u1 = unit_step(t-1)
    u2 = unit_step(t-3)
    u3 = unit_step(t-6)
    
    # Combine to create the piecewise function
    return 2*u1 - 3*(t-3)*u2 + (t+2)*u3

# Calculate function values
y = f(t)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(t, y, 'b-', linewidth=2)
plt.grid(True)
plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
plt.axvline(x=1, color='r', linestyle='--', alpha=0.5)
plt.axvline(x=3, color='r', linestyle='--', alpha=0.5)
plt.axvline(x=6, color='r', linestyle='--', alpha=0.5)
plt.xlabel('t')
plt.ylabel('f(t)')
plt.title('Piecewise Function Represented Using Step Functions')
plt.xlim(-1, 10)
plt.ylim(-10, 10)
plt.annotate('t=1', xy=(1, -0.5), xytext=(1, -2),
             arrowprops=dict(arrowstyle='->'))
plt.annotate('t=3', xy=(3, -0.5), xytext=(3, -3),
             arrowprops=dict(arrowstyle='->'))
plt.annotate('t=6', xy=(6, -0.5), xytext=(6, -4),
             arrowprops=dict(arrowstyle='->'))
plt.tight_layout()
plt.show()
```

## The Dirac's Delta Function

### SymPy Implementation for Delta Functions

```python
import sympy as sp

# Define symbols (a > 0 places the impulse inside the
# integration range of the one-sided transform)
t, s = sp.symbols('t s', real=True)
a = sp.symbols('a', positive=True)

# Define the Dirac delta function
delta_t = sp.DiracDelta(t)
delta_t_a = sp.DiracDelta(t-a)

# Compute Laplace transform of delta function
L_delta = sp.laplace_transform(delta_t, t, s)
L_delta_a = sp.laplace_transform(delta_t_a, t, s)

print("Laplace transform of delta(t):", L_delta[0])
print("Laplace transform of delta(t-a):", L_delta_a[0])

# Verify sifting property
f = sp.Function('f')(t)
integral = sp.integrate(f * delta_t_a, (t, -sp.oo, sp.oo))
print("Sifting property result:", integral)

# Verify relationship with step function
step_derivative = sp.diff(sp.Heaviside(t-a), t)
print("Derivative of step function:", step_derivative)
```

## Useful Theorems for Laplace Transform

### Visualizing Theorems with SymPy

```python
import sympy as sp

# Define symbols
t, s, a, omega = sp.symbols('t s a omega', real=True, positive=True)

# Define a function and its Laplace transform
f = sp.exp(-2*t)
F = sp.laplace_transform(f, t, s)[0]

print("Function:", f)
print("Laplace transform:", F)

# Verify time shifting theorem
f_shifted = sp.exp(-2*(t-a))*sp.Heaviside(t-a)
F_shifted = sp.laplace_transform(f_shifted, t, s)[0]
F_shifted_expected = sp.exp(-a*s)*F
print("\nTime shifting theorem:")
print("L{f(t-a)u(t-a)} =", F_shifted)
print("e^(-as)F(s) =", F_shifted_expected)
print("Equality:", sp.simplify(F_shifted - F_shifted_expected) == 0)

# Verify frequency shifting theorem
f_freq_shifted = sp.exp(a*t)*f
F_freq_shifted = sp.laplace_transform(f_freq_shifted, t, s)[0]
F_freq_shifted_expected = F.subs(s, s-a)
print("\nFrequency shifting theorem:")
print("L{e^(at)f(t)} =", F_freq_shifted)
print("F(s-a) =", F_freq_shifted_expected)
print("Equality:", sp.simplify(F_freq_shifted - F_freq_shifted_expected) == 0)

# Verify time differentiation theorem
f_prime = sp.diff(f, t)
F_prime = sp.laplace_transform(f_prime, t, s)[0]
F_prime_expected = s*F - f.subs(t, 0)
print("\nTime differentiation theorem:")
print("L{f'(t)} =", F_prime)
print("sF(s) - f(0) =", F_prime_expected)
print("Equality:", sp.simplify(F_prime - F_prime_expected) == 0)
```

## Laplace Transforms for Periodic Functions

### SymPy Implementation for Periodic Functions

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Define symbols
t, s = sp.symbols('t s', positive=True)
T_val = 2

# Transform of one period (amplitude 1, on for the first
# half-period), then the periodic-function formula
one_period = sp.integrate(sp.exp(-s*t), (t, 0, sp.Rational(T_val, 2)))
F_s = sp.simplify(one_period / (1 - sp.exp(-s*T_val)))
print("F(s) =", F_s)

# Numerical square wave for plotting
def square_wave(t_arr, T):
    return np.where((t_arr % T) < T/2, 1.0, 0.0)

t_vals = np.linspace(0, 10, 1000)
square_vals = square_wave(t_vals, T_val)

# Evaluate |F(s)| along the real axis s > 0
F_func = sp.lambdify(s, F_s, 'numpy')
s_vals = np.linspace(0.05, 5, 500)
F_vals = np.abs(F_func(s_vals))

# Plot with subplots
fig, axs = plt.subplots(2, 1, figsize=(10, 8))

# Subplot 1: Square wave
axs[0].plot(t_vals, square_vals, 'b-', linewidth=2)
axs[0].set_title('Square Wave with Period $T = 2$')
axs[0].set_xlabel('Time $t$ (s)')
axs[0].set_ylabel('Amplitude $f(t)$')
axs[0].grid(True)
axs[0].set_ylim(-0.1, 1.1)

# Subplot 2: Laplace magnitude
axs[1].plot(s_vals, F_vals, 'r-', linewidth=2)
axs[1].set_title('Magnitude of Laplace Transform of Square Wave')
axs[1].set_xlabel('Real frequency $s$ (1/s)')
axs[1].set_ylabel(r'$|F(s)|$')
axs[1].grid(True)

plt.tight_layout()
plt.show()
```

## Convolution

### Examples of Convolution

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from sympy.integrals.transforms import laplace_transform, inverse_laplace_transform

# Define symbols
t, s, tau, a, b = sp.symbols('t s tau a b', real=True, positive=True)

# Define functions
f_t = sp.exp(-a*t)
g_t = sp.exp(-b*t)

# Method 1: Direct integration
conv_integral = sp.integrate(f_t.subs(t, tau) * g_t.subs(t, t-tau), (tau, 0, t))
conv_result = sp.simplify(conv_integral.subs([(a, 1), (b, 2)]))
print("Convolution by direct integration:")
print(sp.simplify(conv_integral))
print("For a=1, b=2:", conv_result)

# Method 2: Using Laplace transforms
F_s = laplace_transform(f_t, t, s, noconds=True)
G_s = laplace_transform(g_t, t, s, noconds=True)
FG_s = F_s * G_s
conv_laplace = inverse_laplace_transform(FG_s, s, t, noconds=True)
conv_laplace_result = sp.simplify(conv_laplace.subs([(a, 1), (b, 2)]))
print("\nConvolution by Laplace transforms:")
print(sp.simplify(conv_laplace))
print("For a=1, b=2:", conv_laplace_result)

# Plot the functions and their convolution for specific values
def plot_functions_and_convolution(a_val=1, b_val=2, t_max=5):
    t_vals = np.linspace(0, t_max, 1000)
    f_vals = np.exp(-a_val * t_vals)
    g_vals = np.exp(-b_val * t_vals)
    conv_vals = (np.exp(-a_val * t_vals) - np.exp(-b_val * t_vals))/(b_val - a_val)
    
    plt.figure(figsize=(10, 6))
    plt.plot(t_vals, f_vals, 'b-', label=f'f(t) = e^(-{a_val}t)')
    plt.plot(t_vals, g_vals, 'r-', label=f'g(t) = e^(-{b_val}t)')
    plt.plot(t_vals, conv_vals, 'g-', label='(f*g)(t)')
    plt.xlabel('t')
    plt.ylabel('Function Value')
    plt.grid(True)
    plt.legend()
    plt.title('Functions and Their Convolution')
    plt.show()

# Call the plotting function
plot_functions_and_convolution()
```

### Physical Interpretation of Convolution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
def rect_pulse(t, start=0, width=1, height=1):
    """Rectangular pulse function"""
    return height * ((t >= start) & (t < start + width))
def impulse_response(t, decay_rate=1):
    """Exponential decay impulse response"""
    return np.exp(-decay_rate * t) * (t >= 0)
# Create time vector
t = np.linspace(0, 10, 1000)
dt = t[1] - t[0]
# Create input signal (rectangular pulse)
input_signal = rect_pulse(t, start=2, width=4, height=1)
# Create impulse response
h = impulse_response(t, decay_rate=1)
# Compute convolution (system output); the factor dt
# approximates the convolution integral
output_signal = signal.convolve(input_signal, h, mode='full')[:len(t)] * dt
# Plot input, impulse response, output, and an overlay
fig, axs = plt.subplots(4, 1, figsize=(10, 11), sharex=True)
axs[0].plot(t, input_signal, 'b-')
axs[0].set_title('Input Signal (Rectangular Pulse)')
axs[0].set_ylabel('x(t)')
axs[0].grid(True)
axs[1].plot(t, h, 'r-')
axs[1].set_title('Impulse Response (Exponential Decay)')
axs[1].set_ylabel('h(t)')
axs[1].grid(True)
axs[2].plot(t, output_signal, 'g-')
axs[2].set_title('Output Signal (Convolution Result)')
axs[2].set_ylabel('y(t)')
axs[2].grid(True)
axs[3].plot(t, input_signal, 'b-', label='Input')
axs[3].plot(t, h, 'r-', label='Impulse response')
axs[3].plot(t, output_signal, 'g-', label='Output')
axs[3].set_title('Convolution as System Response')
axs[3].set_xlabel('Time t (s)')
axs[3].set_ylabel('Amplitude')
axs[3].grid(True)
axs[3].legend()
plt.tight_layout()
plt.show()
```

### The Hayami Method in Hydrology

#### Derivation from First Principles

**Hayami Kernel Verification in Python**

```python
import sympy as sp
import numpy as np

# Define symbols
t, x_sym, C_sym, k_sym = sp.symbols('t x C k', positive=True)

# Symbolic Hayami kernel
hayami_expr = (x_sym / (2 * sp.sqrt(sp.pi * k_sym * t**3))) * \
              sp.exp(-(x_sym - C_sym*t)**2 / (4 * k_sym*t))

# Lambdify to numeric
kernel_func = sp.lambdify((t, x_sym, C_sym, k_sym), hayami_expr, 'numpy')

# Parameters
x = 20000    # meters
C = 1.0      # m/s
k = 500      # m^2/s

# Time array
times_hours = np.arange(0.1, 30.1, 0.1)
times_seconds = times_hours * 3600

# Compute responses
responses = kernel_func(times_seconds, x, C, k)

# Peak response
peak_idx = np.argmax(responses)
peak_response = responses[peak_idx]
peak_time = times_hours[peak_idx]
print(f"Peak response: {peak_response:.6e} at time: {peak_time:.1f} hours")

# Wave arrival at 1% of peak
threshold = 0.01 * peak_response
arrival_indices = np.where(responses > threshold)[0]
arrival_time = times_hours[arrival_indices[0]] if arrival_indices.size else None
print(f"Wave arrival (1% of peak): {arrival_time:.1f} hours")

# Numerical integration (trapezoidal rule)
integral = np.trapezoid(responses, times_seconds)
print(f"Numerical integral: {integral:.6f}")
```

#### Physical Interpretation

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

def hayami_impulse_response(t, x, C, k):
    """
    Calculate the Hayami impulse response function.
    
    Parameters:
    -----------
    t : array_like
        Time array (seconds)
    x : float
        Distance downstream (meters)
    C : float
        Celerity/wave speed (m/s)
    k : float
        Diffusion coefficient (m2/s)
    
    Returns:
    --------
    h : array_like
        Impulse response values corresponding to t
    """
    # Avoid division by zero by setting very small values to NaN
    t_valid = np.where(t > 0, t, np.nan)

    # Calculate the impulse response
    h = (x / (2 * np.sqrt(np.pi * k * t_valid**3))) * \
        np.exp(-((x - C * t_valid)**2) / (4 * k * t_valid))
    
    # Replace NaN values with zeros (for t=0)
    h = np.nan_to_num(h)
    
    return h

def triangular_hydrograph(t, peak_time=10, peak_flow=100, base_time=30):
    """
    Generate a triangular hydrograph (simplified input flood wave).
    
    Parameters:
    -----------
    t : array_like
        Time array
    peak_time : float
        Time to peak flow (hours)
    peak_flow : float
        Maximum flow rate (m3/s)
    base_time : float
        Total duration of the hydrograph (hours)
    
    Returns:
    --------
    flow : array_like
        Flow values corresponding to t
    """
    flow = np.zeros_like(t)
    
    # Rising limb
    rising_mask = (t >= 0) & (t < peak_time)
    flow[rising_mask] = peak_flow * (t[rising_mask] / peak_time)
    
    # Falling limb
    falling_mask = (t >= peak_time) & (t <= base_time)
    flow[falling_mask] = peak_flow * (1 - (t[falling_mask] - peak_time) / (base_time - peak_time))
    
    return flow

# Time parameters
dt = 0.1  # Time step (hours)
duration = 100  # Simulation duration (hours)
t = np.arange(0, duration, dt)  # Time array (hours)

# Generate upstream hydrograph (input)
upstream_flow = triangular_hydrograph(t, peak_time=10, peak_flow=100, base_time=30)

# Channel parameters
x = 20000  # Distance downstream (m)
C = 1.0  # Celerity (m/s)
k = 500  # Diffusion coefficient (m2/s)

# Convert time to seconds for the impulse response calculation
t_seconds = t * 3600

# Calculate Hayami impulse response
h = hayami_impulse_response(t_seconds, x, C, k)

# Normalize the impulse response for visualization
h_normalized = h / np.max(h)

# Perform convolution to get downstream hydrograph
# Note: We multiply by dt (in seconds) to account for the discrete nature of convolution
downstream_flow = signal.convolve(upstream_flow, h * dt * 3600, mode='full')[:len(t)]

# Plot results
plt.figure(figsize=(12, 10))

# Plot upstream and downstream hydrographs
plt.subplot(3, 1, 1)
plt.plot(t, upstream_flow, 'b-', linewidth=2, label='Upstream Flow')
plt.plot(t, downstream_flow, 'r-', linewidth=2, label='Downstream Flow')
plt.grid(True)
plt.legend()
plt.title('Hayami Method: Flood Wave Routing')
plt.xlabel('Time (hours)')
plt.ylabel('Discharge (m$^3$/s)')

# Plot impulse response
plt.subplot(3, 1, 2)
plt.plot(t, h, 'g-', linewidth=2)
plt.grid(True)
plt.title('Hayami Impulse Response Function')
plt.xlabel('Time (hours)')
plt.ylabel('Response (1/s)')

# Plot both on same axes for comparison
plt.subplot(3, 1, 3)
plt.plot(t, upstream_flow / np.max(upstream_flow), 'b-', linewidth=2, label='Normalized Upstream Flow')
plt.plot(t, downstream_flow / np.max(downstream_flow), 'r-', linewidth=2, label='Normalized Downstream Flow')
plt.plot(t, h_normalized, 'g--', linewidth=1.5, label='Normalized Impulse Response')
plt.grid(True)
plt.legend()
plt.title('Comparison of Normalized Flows')
plt.xlabel('Time (hours)')
plt.ylabel('Normalized Value')

plt.tight_layout()
plt.show()

# Calculate and print key hydrograph characteristics
upstream_peak = np.max(upstream_flow)
downstream_peak = np.max(downstream_flow)
upstream_peak_time = t[np.argmax(upstream_flow)]
downstream_peak_time = t[np.argmax(downstream_flow)]
travel_time = downstream_peak_time - upstream_peak_time
attenuation_ratio = downstream_peak / upstream_peak

print(f"Upstream peak flow: {upstream_peak:.2f} m3/s at t = {upstream_peak_time:.2f} hours")
print(f"Downstream peak flow: {downstream_peak:.2f} m3/s at t = {downstream_peak_time:.2f} hours")
print(f"Wave travel time: {travel_time:.2f} hours")
print(f"Peak attenuation ratio: {attenuation_ratio:.2f}")
```

## Solutions of Differential Equations with Constant and Variable Coefficients

### Differential Equations with Constant Coefficients

```python
import sympy as sp
from sympy.integrals.transforms import laplace_transform, inverse_laplace_transform
import numpy as np
import matplotlib.pyplot as plt

# Define symbols
t = sp.symbols('t', positive=True)
s = sp.symbols('s', real=True)
y = sp.Function('y')(t)

# Define the differential equation
diffeq = sp.Eq(sp.diff(y, t, 2) + 4*sp.diff(y, t) + 3*y, 2*sp.exp(-t))

# Define initial conditions
ics = {y.subs(t, 0): 1, sp.diff(y, t).subs(t, 0): 0}

# Print the differential equation and initial conditions
print("Differential equation:")
print(diffeq)
print("\nInitial conditions:")
for ic, val in ics.items():
    print(f"{ic} = {val}")

# Apply Laplace transform
Y = sp.Function('Y')(s)
y_0 = ics[y.subs(t, 0)]
y_1 = ics[sp.diff(y, t).subs(t, 0)]

# Manual solution using Laplace transform
print("\nApplying Laplace transform:")
print(f"L{{y''}} = s^2*Y(s) - s*y(0) - y'(0) = s^2*Y(s) - {s*y_0} - {y_1}")
print(f"L{{y'}} = s*Y(s) - y(0) = s*Y(s) - {y_0}")
print(f"L{{y}} = Y(s)")
print(f"L{{2*e^(-t)}} = 2/(s+1)")

# Formulate the transformed equation
laplace_eq = sp.Eq(
    s**2 * Y - s*y_0 - y_1 + 4*(s*Y - y_0) + 3*Y,
    2/(s+1)
)
print("\nTransformed equation:")
print(laplace_eq)

# Solve for Y(s)
Y_sol = sp.solve(laplace_eq, Y)[0]
print("\nSolution for Y(s):")
print(Y_sol)

# Simplify the expression
Y_sol_simplified = sp.simplify(Y_sol)
print("\nSimplified Y(s):")
print(Y_sol_simplified)

# Decompose into partial fractions
Y_sol_pf = sp.apart(Y_sol_simplified, s)
print("\nPartial fraction decomposition:")
print(Y_sol_pf)

# Find inverse Laplace transform
y_sol = inverse_laplace_transform(Y_sol_pf, s, t)
print("\nSolution y(t):")
print(y_sol)
print("\nSimplified solution:")
print(sp.simplify(y_sol))

# Plot the solution
def plot_solution(t_max=5):
    t_vals = np.linspace(0, t_max, 1000)
    y_vals = [(1 + t_val) * np.exp(-t_val) for t_val in t_vals]
    
    plt.figure(figsize=(10, 6))
    plt.plot(t_vals, y_vals, 'b-', linewidth=2)
    plt.grid(True)
    plt.xlabel('t')
    plt.ylabel('y(t)')
    plt.title(r'Solution: $y(t) = e^{-t}(1 + t)$')
    plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
    plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
    plt.tight_layout()
    plt.show()

# Call plotting function
plot_solution()
```

### Differential Equations with Variable Coefficients

```python
import numpy as np
import matplotlib.pyplot as plt

# Solution y(t) = 1/(4t^3) + 7t/4 and its two terms
t_vals = np.linspace(0.2, 3, 1000)
y_vals = 1/(4*t_vals**3) + 7*t_vals/4
y_term1 = 1/(4*t_vals**3)
y_term2 = 7*t_vals/4

plt.figure(figsize=(10, 6))
plt.plot(t_vals, y_vals, 'b-', linewidth=2, label='Complete solution')
plt.plot(t_vals, y_term1, 'r--', label=r'Term $1/(4t^3)$')
plt.plot(t_vals, y_term2, 'g--', label=r'Term $7t/4$')
plt.scatter([1], [2], color='black', s=50, label='Initial point (1,2)')

plt.grid(True)
plt.xlabel('t')
plt.ylabel('y(t)')
plt.legend()
plt.xlim(0.2, 3)
plt.ylim(0, 6)
plt.tight_layout()
plt.show()
```

### Applications in Mechanical Systems

#### Case (b): Damped Motion

```python
import numpy as np
import matplotlib.pyplot as plt

def analyze_particle_motion(t_max=6.0, n_points=1000):
    """
    Analyze and visualize the motion of a particle attracted to the origin,
    with and without damping.

    Parameters:
    -----------
    t_max : float
        Maximum time for simulation (s)
    n_points : int
        Number of time points for simulation
    """
    # System parameters
    m = 2.0    # Mass (kg)
    k = 8.0    # Spring constant (N/m)
    c = 10.0   # Damping coefficient (N s/m)
    
    # Initial conditions
    x0 = 10.0  # Initial position (m)
    v0 = 0.0   # Initial velocity (m/s)
    
    # Calculate key system parameters
    omega_n = np.sqrt(k/m)  # Natural frequency (rad/s)
    damping_ratio = c / (2 * np.sqrt(m * k))  # Damping ratio (dimensionless)
    
    # Time array
    t = np.linspace(0, t_max, n_points)
    
    # Case (a): No damping (c = 0)
    # Solution: x(t) = x0 * cos(omega_n * t)
    x_undamped = x0 * np.cos(omega_n * t)
    v_undamped = -x0 * omega_n * np.sin(omega_n * t)
    
    # Case (b): With damping (overdamped system since zeta > 1)
    # For an overdamped system, we calculate the roots of the characteristic equation
    s1 = -omega_n * (damping_ratio + np.sqrt(damping_ratio**2 - 1))
    s2 = -omega_n * (damping_ratio - np.sqrt(damping_ratio**2 - 1))
    
    # Constants for the solution based on initial conditions
    C1 = x0 * (s2) / (s2 - s1)
    C2 = x0 * (-s1) / (s2 - s1)
    
    # Solution for overdamped case
    x_damped = C1 * np.exp(s1 * t) + C2 * np.exp(s2 * t)
    v_damped = C1 * s1 * np.exp(s1 * t) + C2 * s2 * np.exp(s2 * t)
    
    # Energy calculations
    # Kinetic energy: KE = 0.5 * m * v^2
    KE_undamped = 0.5 * m * v_undamped**2
    KE_damped = 0.5 * m * v_damped**2
    
    # Potential energy: PE = 0.5 * k * x^2
    PE_undamped = 0.5 * k * x_undamped**2
    PE_damped = 0.5 * k * x_damped**2
    
    # Total energy: E = KE + PE
    E_undamped = KE_undamped + PE_undamped
    E_damped = KE_damped + PE_damped
    
    # Create plots
    plt.figure(figsize=(12, 10))
    
    # Plot position
    plt.subplot(3, 1, 1)
    plt.plot(t, x_undamped, 'b-', label='Undamped')
    plt.plot(t, x_damped, 'r-', label='Damped')
    plt.axhline(y=0, color='k', linestyle='--', alpha=0.3)
    plt.grid(True)
    plt.xlabel('Time (s)')
    plt.ylabel('Position (m)')
    plt.title('Particle Position vs Time')
    plt.legend()
    
    # Plot velocity
    plt.subplot(3, 1, 2)
    plt.plot(t, v_undamped, 'b-', label='Undamped')
    plt.plot(t, v_damped, 'r-', label='Damped')
    plt.axhline(y=0, color='k', linestyle='--', alpha=0.3)
    plt.grid(True)
    plt.xlabel('Time (s)')
    plt.ylabel('Velocity (m/s)')
    plt.title('Particle Velocity vs Time')
    plt.legend()
    
    # Plot energy
    plt.subplot(3, 1, 3)
    plt.plot(t, E_undamped, 'b-', label='Undamped Total Energy')
    plt.plot(t, E_damped, 'r-', label='Damped Total Energy')
    plt.plot(t, KE_undamped, 'b--', alpha=0.5, label='Undamped Kinetic Energy')
    plt.plot(t, PE_undamped, 'b:', alpha=0.5, label='Undamped Potential Energy')
    plt.grid(True)
    plt.xlabel('Time (s)')
    plt.ylabel('Energy (J)')
    plt.title('Energy vs Time')
    plt.legend()
    
    plt.tight_layout()
    
    # Phase portrait (position vs velocity)
    plt.figure(figsize=(10, 8))
    plt.plot(x_undamped, v_undamped, 'b-', label='Undamped')
    plt.plot(x_damped, v_damped, 'r-', label='Damped')
    plt.plot(x0, v0, 'ko', label='Initial State')
    plt.grid(True)
    plt.xlabel('Position (m)')
    plt.ylabel('Velocity (m/s)')
    plt.title('Phase Portrait')
    plt.legend()
    plt.axis('equal')
    
    # Display system characteristics
    print("System Characteristics:")
    print(f"Mass: {m:.1f} kg")
    print(f"Spring Constant: {k:.1f} N/m")
    print(f"Damping Coefficient: {c:.1f} N s/m")
    print(f"Natural Frequency: {omega_n:.2f} rad/s")
    print(f"Natural Period: {2*np.pi/omega_n:.4f} s")
    print(f"Damping Ratio: {damping_ratio:.4f}")
    
    # Calculate the time for the amplitude to decay to 1% of initial value
    if damping_ratio < 1:  # Underdamped
        decay_time = -np.log(0.01) / (damping_ratio * omega_n)
        print(f"Time to decay to 1% (undamped): infinite (no decay)")
        print(f"Time to decay to 1% (damped): {decay_time:.4f} s")
    elif damping_ratio == 1:  # Critically damped
        decay_time = -np.log(0.01) / omega_n
        print(f"Time to decay to 1% (damped): {decay_time:.4f} s")
    else:  # Overdamped
        decay_time = -np.log(0.01) / (omega_n * (damping_ratio - np.sqrt(damping_ratio**2 - 1)))
        print(f"Time to decay to 1% (damped): {decay_time:.4f} s")
    
    plt.show()

    # Return the time and position arrays for further analysis
    return t, x_undamped, x_damped

# Run the analysis
t, x_undamped, x_damped = analyze_particle_motion(t_max=6.0)
```

## The Z-Transform: Discrete Analog of Laplace Transform

### Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Example: First-order discrete system
# y[n] = 0.8*y[n-1] + 0.5*u[n]
# Rearranging: y[n] - 0.8*y[n-1] = 0.5*u[n]

# In the Z-domain: Y(z) - 0.8*z^{-1}*Y(z) = 0.5*U(z)
# Transfer function: H(z) = Y(z)/U(z) = 0.5/(1 - 0.8*z^{-1})

# Define the discrete system using scipy
# Numerator and denominator coefficients
# H(z) = 0.5/(1 - 0.8*z^{-1}) = 0.5*z/(z - 0.8)
b = [0.5]       # Numerator coefficients
a = [1, -0.8]   # Denominator coefficients

# Create discrete-time LTI system
sys = signal.TransferFunction(b, a, dt=1.0)

print("Discrete-time transfer function:")
print(f"H(z) = {b[0]}/({a[0]} - {-a[1]}*z^-1)")

# Analyze pole-zero locations
poles = sys.poles
zeros = sys.zeros

print(f"\nPoles: {poles}")
print(f"Zeros: {zeros}")
print(f"System is stable: {all(np.abs(poles) < 1)}")

# Pole-zero diagram
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4))

# Pole-zero plot
theta = np.linspace(0, 2*np.pi, 100)
ax1.plot(np.cos(theta), np.sin(theta), 'k--', alpha=0.3, linewidth=1)
ax1.plot(np.real(zeros), np.imag(zeros), 'bo', markersize=10,
         markerfacecolor='none', markeredgewidth=2, label='Zeros')
ax1.plot(np.real(poles), np.imag(poles), 'rx', markersize=12,
         markeredgewidth=2, label='Poles')
ax1.axhline(y=0, color='k', linewidth=0.5)
ax1.axvline(x=0, color='k', linewidth=0.5)
ax1.set_xlabel('Real', fontsize=11)
ax1.set_ylabel('Imaginary', fontsize=11)
ax1.set_title('Pole-Zero Diagram in Z-Plane', fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.axis('equal')
ax1.set_xlim([-1.5, 1.5])
ax1.set_ylim([-1.5, 1.5])
ax1.legend(fontsize=10)

# Impulse response
n_samples = 50
t = np.arange(n_samples)
impulse_response = np.zeros(n_samples)
impulse_response[0] = b[0]
for n in range(1, n_samples):
    impulse_response[n] = 0.8*impulse_response[n-1]

ax2.stem(t, impulse_response, basefmt=' ')
ax2.set_xlabel('n', fontsize=11)
ax2.set_ylabel('h[n]', fontsize=11)
ax2.set_title('Impulse Response', fontsize=12)
ax2.grid(True, alpha=0.3)

# Frequency response (z = e^{jw} on the unit circle):
# H(e^{jw}) = b0 / (1 + a1*e^{-jw}) with a1 = -0.8
w = np.linspace(0, np.pi, 1000)
H = b[0] / (1 + a[1]*np.exp(-1j*w))
magnitude = np.abs(H)

ax3.plot(w/np.pi, 20*np.log10(magnitude + 1e-10), 'b-', linewidth=2)
ax3.set_xlabel('Normalized Frequency (pi rad/sample)', fontsize=11)
ax3.set_ylabel('Magnitude (dB)', fontsize=11)
ax3.set_title('Frequency Response', fontsize=12)
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('figures/chapter_5_auto_21.png', dpi=200)
plt.close()

print("\nZ-transform analysis plotted")

# Example: Digital low-pass filter
# Design using bilinear transformation
# Continuous prototype: H_c(s) = 1/(s + 1)
# Sample period: T = 0.1 seconds

T = 0.1
# Bilinear transformation: s = 2(z-1)/(T(z+1))
# After substitution and algebra: H(z) = T/(T + 2) * (z + 1)/(z - (2-T)/(2+T))

num = T * np.array([1, 1])
den = (T + 2) * np.array([1, -(2-T)/(2+T)])
H_digital = signal.TransferFunction(num/den[0], den/den[0], dt=T)

print(f"\nDigital filter (bilinear from 1/(s+1)):")
print(f"H(z) = {H_digital.num} / {H_digital.den}")
```

## State-Space Methods and Laplace Transforms

### Python Implementation with scipy.signal

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal, integrate

# Example: Two-DOF (degrees of freedom) mass-spring system
# Mass 1 is anchored to ground by (k1, c1) and coupled to
# mass 2 by (k2, c2):
# m1*x1'' + c1*x1' + k1*x1 + k2*(x1-x2) + c2*(x1'-x2') = F1
# m2*x2'' + k2*(x2-x1) + c2*(x2'-x1') = F2

# State vector: [x1, x1', x2, x2']
# Define system matrices

m1, m2 = 1.0, 1.0     # masses
k1, k2 = 10.0, 5.0    # spring constants
c1, c2 = 0.5, 0.2     # damping coefficients

# State-space matrices
# dx1/dt = x2
# dx2/dt = (F1 - c1*x2 - k1*x1 + c2*(x4-x2) + k2*(x3-x1))/m1
# dx3/dt = x4
# dx4/dt = (F2 - c2*x4 - k2*x3 + c2*(x2-x4) + k2*(x1-x3))/m2

A = np.array([
    [0, 1, 0, 0],
    [-(k1+k2)/m1, -(c1+c2)/m1, k2/m1, c2/m1],
    [0, 0, 0, 1],
    [k2/m2, c2/m2, -k2/m2, -c2/m2]
])

B = np.array([
    [0, 0],
    [1/m1, 0],
    [0, 0],
    [0, 1/m2]
])

C = np.eye(4)  # Observe all states
D = np.zeros((4, 2))

# Create state-space system
sys = signal.StateSpace(A, B, C, D)

# Check controllability
n = A.shape[0]
P = np.hstack([np.linalg.matrix_power(A, i) @ B
               for i in range(n)])
controllability_rank = np.linalg.matrix_rank(P)
is_controllable = controllability_rank == n

print(f"System Controllability:")
print(f"Controllability matrix rank: {controllability_rank}/{n}")
print(f"System is controllable: {is_controllable}\n")

# Check observability
Q = np.vstack([C @ np.linalg.matrix_power(A, i)
               for i in range(n)])
observability_rank = np.linalg.matrix_rank(Q)
is_observable = observability_rank == n

print(f"System Observability:")
print(f"Observability matrix rank: {observability_rank}/{n}")
print(f"System is observable: {is_observable}\n")

# Eigenvalues (poles) of the system
eigenvalues = np.linalg.eigvals(A)
print(f"System poles (eigenvalues): {eigenvalues}")
is_stable = all(np.real(eigenvalues) < 0)
print(f"System is stable: {is_stable}\n")

# Simulate response to a force applied to mass 1
duration = 10.0
t_interp = np.linspace(0, duration, 1000)
u_force = np.zeros((len(t_interp), 2))
u_force[100:300, 0] = 1.0  # rectangular force pulse on mass 1

def u_func_interp(time):
    idx = min(int(time / duration * len(u_force)), len(u_force) - 1)
    return u_force[idx]

# Integrate the state equations
def state_equations(x, time):
    u = u_func_interp(time)
    return A @ x + B @ u

# Initial conditions
x0 = np.zeros(4)

# Solve ODE
sol = integrate.odeint(state_equations, x0, t_interp)

# Plot results
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# State trajectories
ax = axes[0, 0]
ax.plot(t_interp, sol[:, 0], 'b-', linewidth=2, label='$x_1$ (pos)')
ax.plot(t_interp, sol[:, 2], 'r-', linewidth=2, label='$x_2$ (pos)')
ax.set_xlabel('Time (s)', fontsize=11)
ax.set_ylabel('Position (m)', fontsize=11)
ax.set_title('Mass Positions', fontsize=12)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)

# Velocities
ax = axes[0, 1]
ax.plot(t_interp, sol[:, 1], 'b-', linewidth=2, label='$x_1\'$')
ax.plot(t_interp, sol[:, 3], 'r-', linewidth=2, label='$x_2\'$')
ax.set_xlabel('Time (s)', fontsize=11)
ax.set_ylabel('Velocity (m/s)', fontsize=11)
ax.set_title('Mass Velocities', fontsize=12)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)

# Input force
ax = axes[1, 0]
ax.plot(t_interp, u_force[:, 0], 'g-', linewidth=2)
ax.set_xlabel('Time (s)', fontsize=11)
ax.set_ylabel('Force (N)', fontsize=11)
ax.set_title('Input Force on Mass 1', fontsize=12)
ax.grid(True, alpha=0.3)

# System eigenvalues
ax = axes[1, 1]
ax.plot(np.real(eigenvalues), np.imag(eigenvalues), 'kx',
        markersize=12, markeredgewidth=2)
ax.axvline(x=0, color='k', linewidth=0.8)
ax.axhline(y=0, color='k', linewidth=0.8)
ax.set_xlabel('Real Part', fontsize=11)
ax.set_ylabel('Imaginary Part', fontsize=11)
ax.set_title('Pole Locations (Eigenvalues)', fontsize=12)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('state_space_analysis.png', dpi=200)
plt.close()

print("State-space analysis completed and plotted")
```
