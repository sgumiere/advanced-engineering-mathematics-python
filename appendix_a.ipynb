## Python Quick Reference

### Essential NumPy Operations

**NumPy quick reference**

```python
import numpy as np

# Array creation
A = np.array([[1, 2], [3, 4]])
b = np.array([5, 6])
grid = np.linspace(0, 1, 100)
I = np.eye(3)
Z = np.zeros((3, 3))

# Linear algebra
x = np.linalg.solve(A, b)           # Solve Ax = b
evals, evecs = np.linalg.eig(A)     # Eigenvalues
U, S, Vt = np.linalg.svd(A)         # SVD
det_A = np.linalg.det(A)            # Determinant
inv_A = np.linalg.inv(A)            # Inverse
norm_A = np.linalg.norm(A)          # Frobenius norm
cond_A = np.linalg.cond(A)          # Condition number
rank_A = np.linalg.matrix_rank(A)   # Rank
```

### Essential SciPy Operations

**SciPy quick reference**

```python
from scipy import integrate, optimize, linalg, interpolate

# Integration
result, err = integrate.quad(lambda x: x**2, 0, 1)

# ODE solving
from scipy.integrate import solve_ivp
sol = solve_ivp(lambda t, y: -y, [0, 5], [1])

# Root finding
root = optimize.fsolve(lambda x: x**2 - 2, 1.5)

# Optimization
result = optimize.minimize(lambda x: x[0]**2 + x[1]**2,
                          x0=[1, 1], method='BFGS')

# Interpolation
import numpy as np
x_data = np.linspace(0, 1, 10)
y_data = np.sin(2 * np.pi * x_data)
f = interpolate.interp1d(x_data, y_data, kind='cubic')

# Sparse matrices
from scipy.sparse import csr_matrix, linalg as splinalg
A = np.array([[4.0, 1.0], [1.0, 3.0]])
b = np.array([1.0, 2.0])
A_sparse = csr_matrix(A)
x = splinalg.spsolve(A_sparse, b)
```

### Essential SymPy Operations

**SymPy quick reference**

```python
import sympy as sp

x, y, t, s = sp.symbols('x y t s')

# Calculus
sp.diff(sp.sin(x)*sp.exp(x), x)      # Derivative
sp.integrate(sp.cos(x), x)             # Integral
sp.limit(sp.sin(x)/x, x, 0)           # Limit
sp.series(sp.exp(x), x, 0, 5)         # Taylor series

# Linear algebra
M = sp.Matrix([[1, 2], [3, 4]])
M.eigenvals()                           # Eigenvalues
M.eigenvects()                          # Eigenvectors
M.det()                                 # Determinant

# ODEs
y = sp.Function('y')
ode = sp.Eq(y(x).diff(x, 2) + y(x), 0)
sp.dsolve(ode, y(x))

# Laplace transforms
sp.laplace_transform(sp.sin(t), t, s)
sp.inverse_laplace_transform(1/(s**2+1), s, t)
```

### Essential PyTorch Operations for PINNs

**PyTorch quick reference for scientific computing**

```python
import torch
import torch.nn as nn

# Tensor with gradient tracking
x = torch.linspace(0, 1, 100, requires_grad=True)

# Automatic differentiation
y = torch.sin(x)
dy_dx = torch.autograd.grad(y.sum(), x,
                             create_graph=True)[0]

# Simple neural network
model = nn.Sequential(
    nn.Linear(2, 64),
    nn.Tanh(),
    nn.Linear(64, 64),
    nn.Tanh(),
    nn.Linear(64, 1)
)

# Training loop (here: fit model to a target function)
x_train = torch.rand(200, 2)
y_train = torch.sin(x_train[:, :1]) * x_train[:, 1:]
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
for epoch in range(1000):
    optimizer.zero_grad()
    loss = ((model(x_train) - y_train) ** 2).mean()
    loss.backward()
    optimizer.step()
```
