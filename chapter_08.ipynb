## Vectors in 2-Space and 3-Space

### Vector Representation in Multiple Coordinate Systems

**Basic Vector Operations in Python**

```python
import numpy as np

# Define vectors
a = np.array([3, 2, 1])
b = np.array([1, 4, 2])

# Vector magnitude
a_magnitude = np.linalg.norm(a)
print(f"Magnitude of a: {a_magnitude:.4f}")

# Unit vector
a_unit = a / a_magnitude
print(f"Unit vector of a: {a_unit}")
print(f"Magnitude of unit vector: {np.linalg.norm(a_unit):.4f}")

# Vector addition
c = a + b
print(f"a + b = {c}")

# Vector subtraction
d = a - b
print(f"a - b = {d}")

# Scalar multiplication
e = 2 * a
print(f"2a = {e}")

# Position and displacement vectors
r1 = np.array([1.0, 2.0, 3.0])
r2 = np.array([4.0, 5.0, 6.0])
displacement = r2 - r1
print(f"\nPosition vector 1: {r1}")
print(f"Position vector 2: {r2}")
print(f"Displacement vector: {displacement}")
print(f"Displacement magnitude: {np.linalg.norm(displacement):.4f}")
```

## Inner Product (Dot Product)

### Angle Between Vectors

**Dot Product, Projection, and Angle Calculations**

```python
import numpy as np

# Define vectors
a = np.array([4, 3, 0])
b = np.array([5, 1, 0])

# Calculate dot product
dot_product = np.dot(a, b)
print(f"a*b = {dot_product}")

# Calculate magnitudes
mag_a = np.linalg.norm(a)
mag_b = np.linalg.norm(b)
print(f"|a| = {mag_a:.4f}")
print(f"|b| = {mag_b:.4f}")

# Calculate angle between vectors
cos_theta = dot_product / (mag_a * mag_b)
theta_rad = np.arccos(cos_theta)
theta_deg = np.degrees(theta_rad)
print(f"Angle between vectors: {theta_deg:.2f} degrees")

# Calculate scalar projection of a onto b
proj_scalar = dot_product / mag_b
print(f"Scalar projection of a onto b: {proj_scalar:.4f}")

# Calculate vector projection of a onto b
proj_vector = (dot_product / (mag_b**2)) * b
print(f"Vector projection of a onto b: {proj_vector}")

# Verify projection properties
# The projection should be parallel to b
proj_mag = np.linalg.norm(proj_vector)
b_unit = b / mag_b
proj_unit = proj_vector / proj_mag if proj_mag > 0 else np.array([0, 0, 0])
print(f"\nProjection magnitude: {proj_mag:.4f}")
print(f"Unit vector of projection: {proj_unit}")
print(f"Unit vector of b: {b_unit}")

# Check orthogonality of perpendicular component
perp = a - proj_vector
print(f"\nPerpendicular component: {perp}")
print(f"proj*perp = {np.dot(proj_vector, perp):.6f} (should be ~0)")

# Check for orthogonal vectors
c = np.array([3, -4, 0])
dot_ac = np.dot(a, c)
print(f"\na*c = {dot_ac} (vectors are orthogonal: {abs(dot_ac) < 1e-10})")
```

### Engineering Application: Work in Mechanical Systems

**Calculating Work in a Mechanical System**

```python
import numpy as np

# Force vector (in Newtons)
F = np.array([50, 30, 20])

# Displacement vector (in meters)
d = np.array([2, 1, 0])

# Calculate work done
work = np.dot(F, d)
print(f"Force: {F} N")
print(f"Displacement: {d} m")
print(f"Work done: {work} J")

# Calculate the angle between force and displacement
mag_F = np.linalg.norm(F)
mag_d = np.linalg.norm(d)
cos_theta = work / (mag_F * mag_d)
theta_rad = np.arccos(cos_theta)
theta_deg = np.degrees(theta_rad)
print(f"Angle between force and displacement: {theta_deg:.2f} degrees")

# Calculate the component of force in the direction of displacement
F_component = work / mag_d
print(f"Component of force in displacement direction: {F_component:.2f} N")

# Calculate power (work per unit time) if displacement takes 5 seconds
time = 5.0  # seconds
power = work / time
print(f"\nPower (average): {power:.2f} W")

# Work done by friction if friction force is opposite to displacement
# Assuming friction coefficient mu = 0.1 and normal force N = 100 N
mu = 0.1
N = 100.0
F_friction = -mu * N * (d / np.linalg.norm(d))  # Friction opposes motion
W_friction = np.dot(F_friction, d)
print(f"\nFriction force: {F_friction} N")
print(f"Work by friction: {W_friction:.2f} J (negative means energy dissipation)")
```

## Vector Product (Cross Product)

### Triple Products

**Cross Product Calculation and Properties**

```python
import numpy as np

# Define vectors
a = np.array([3, 1, 0], dtype=float)
b = np.array([1, 3, 0], dtype=float)
c = np.array([0, 0, 2], dtype=float)

# Calculate cross product
cross_product = np.cross(a, b)
print(f"a = {a}")
print(f"b = {b}")
print(f"a x b = {cross_product}")

# Verify anticommutativity
cross_ba = np.cross(b, a)
print(f"\nb x a = {cross_ba}")
print(f"a x b = -b x a? {np.allclose(cross_product, -cross_ba)}")

# Calculate magnitude (area of parallelogram)
area = np.linalg.norm(cross_product)
print(f"\nArea of parallelogram: {area:.4f}")

# Verify orthogonality to both vectors
dot_a_cross = np.dot(a, cross_product)
dot_b_cross = np.dot(b, cross_product)
print(f"\n(a x b)*a = {dot_a_cross:.10f} (should be ~0)")
print(f"(a x b)*b = {dot_b_cross:.10f} (should be ~0)")

# Cross product with itself
self_cross = np.cross(a, a)
print(f"\na x a = {self_cross} (should be [0,0,0])")

# Scalar triple product (volume of parallelepiped)
scalar_triple = np.dot(a, np.cross(b, c))
print(f"\nScalar triple product a*(b x c) = {scalar_triple:.4f}")
print(f"This is the volume of the parallelepiped formed by a, b, c")

# Compute volume using determinant
volume_det = np.linalg.det(np.array([a, b, c]))
print(f"Volume using determinant: {abs(volume_det):.4f}")

# Vector triple product (BAC-CAB rule)
bac = np.dot(a, c) * b
cab = np.dot(a, b) * c
triple_vector = bac - cab
cross_of_cross = np.cross(a, np.cross(b, c))
print(f"\nVector triple product a x (b x c) = {cross_of_cross}")
print(f"Using BAC-CAB rule: {triple_vector}")
print(f"Match? {np.allclose(cross_of_cross, triple_vector)}")
```

### Engineering Application: Moment of Force (Torque)

**Torque Calculation in Rotational Mechanics**

```python
import numpy as np

# Position vector from pivot to point of force application (in meters)
r = np.array([0.5, 0.3, 0.0])

# Force vector (in Newtons)
F = np.array([0.0, 0.0, -10.0])  # Downward force

# Calculate torque
torque = np.cross(r, F)
print(f"Position vector r: {r} m")
print(f"Force vector F: {F} N")
print(f"Torque tau = r x F: {torque} N*m")

# Calculate magnitude of torque
torque_magnitude = np.linalg.norm(torque)
print(f"\nMagnitude of torque: {torque_magnitude:.4f} N*m")

# Calculate the perpendicular distance (lever arm)
# |tau| = |F| * d, where d is the perpendicular distance
lever_arm = torque_magnitude / np.linalg.norm(F)
print(f"Effective lever arm: {lever_arm:.4f} m")

# Verify: torque magnitude = |F| * |r| * sin(theta)
angle_rad = np.arccos(np.dot(r, F) / (np.linalg.norm(r) * np.linalg.norm(F)))
angle_deg = np.degrees(angle_rad)
expected_torque = np.linalg.norm(F) * np.linalg.norm(r) * np.sin(angle_rad)
print(f"\nAngle between r and F: {angle_deg:.2f} degrees")
print(f"Expected torque magnitude: {expected_torque:.4f} N*m")
print(f"Match? {np.isclose(torque_magnitude, expected_torque)}")

# Multiple forces on a rigid body
r1 = np.array([1.0, 0.0, 0.0])
F1 = np.array([0.0, 10.0, 0.0])
torque1 = np.cross(r1, F1)

r2 = np.array([0.0, 1.0, 0.0])
F2 = np.array([-10.0, 0.0, 0.0])
torque2 = np.cross(r2, F2)

total_torque = torque1 + torque2
print(f"\nTotal torque from two forces: {total_torque} N*m")
print(f"Magnitude: {np.linalg.norm(total_torque):.4f} N*m")
```

## Vector and Scalar Functions and Fields

### Scalar and Vector Fields

**Parametric Curves, Velocity, and Curvature Computation**

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a parametric curve: helix
def helix(t):
    """Parametric helix: (cos(t), sin(t), t/2pi)"""
    return np.array([np.cos(t), np.sin(t), t/(2*np.pi)])

def helix_derivative(t):
    """Velocity vector"""
    return np.array([-np.sin(t), np.cos(t), 1/(2*np.pi)])

def helix_second_derivative(t):
    """Acceleration vector"""
    return np.array([-np.cos(t), -np.sin(t), 0])

# Compute properties at t = pi/2
t_eval = np.pi / 2
r = helix(t_eval)
v = helix_derivative(t_eval)
a = helix_second_derivative(t_eval)

print("Helix at t = pi/2:")
print(f"Position r(t): {r}")
print(f"Velocity v(t): {v}")
print(f"Speed |v(t)|: {np.linalg.norm(v):.4f}")
print(f"Acceleration a(t): {a}")

# Compute arc length from t=0 to t=2pi
from scipy.integrate import quad

def speed(t):
    v = helix_derivative(t)
    return np.linalg.norm(v)

arc_length, _ = quad(speed, 0, 2*np.pi)
print(f"\nArc length from t=0 to t=2pi: {arc_length:.4f}")

# Compute curvature at t
cross_prod = np.cross(v, a)
curvature = np.linalg.norm(cross_prod) / (np.linalg.norm(v)**3)
print(f"Curvature kappa at t = pi/2: {curvature:.4f}")
print(f"Radius of curvature rho = 1/kappa: {1/curvature:.4f}")

# Plot the helix
t_vals = np.linspace(0, 4*np.pi, 200)
r_vals = np.array([helix(t) for t in t_vals])

fig = plt.figure(figsize=(12, 4))

# 3D curve
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot(r_vals[:, 0], r_vals[:, 1], r_vals[:, 2], 'b-', linewidth=2)
ax1.scatter([r[0]], [r[1]], [r[2]], color='red', s=100, label=f'At t=pi/2')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('z')
ax1.set_title('Helix Curve')
ax1.legend()

# Speed along the curve
ax2 = fig.add_subplot(122)
speeds = np.array([np.linalg.norm(helix_derivative(t)) for t in t_vals])
ax2.plot(t_vals, speeds, 'g-', linewidth=2)
ax2.axvline(t_eval, color='red', linestyle='--', label=f't=pi/2')
ax2.set_xlabel('Parameter t')
ax2.set_ylabel('Speed |v(t)|')
ax2.set_title('Speed Along Helix')
ax2.grid(True)
ax2.legend()

plt.tight_layout()
plt.show()
```

### Vector Field Visualization

**Visualization of Scalar and Vector Fields**

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

# Create a grid
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)

# Define a scalar field (temperature)
T = 100 - 5*(X**2 + Y**2)  # Temperature decreases with distance from origin

# Define a vector field (temperature gradient, or heat flux direction)
dTdx = -10*X
dTdy = -10*Y

# Plot scalar field (temperature contours) and vector field
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
contour = plt.contourf(X, Y, T, 20, cmap=cm.coolwarm)
plt.colorbar(contour, label='Temperature (deg C)')
plt.title('Temperature Scalar Field')
plt.xlabel('x (m)')
plt.ylabel('y (m)')

# Plot vector field (temperature gradient/heat flux)
plt.subplot(1, 2, 2)
k = 50  # Thermal conductivity (W/(m*K))
qx = -k * dTdx  # Heat flux components (Fourier's law)
qy = -k * dTdy

plt.quiver(X, Y, qx, qy, color='blue', alpha=0.7)
plt.title('Heat Flux Vector Field (W/m$^2$)')
plt.xlabel('x (m)')
plt.ylabel('y (m)')

# Add streamlines
plt.streamplot(X, Y, qx, qy, color='red', linewidth=0.8, density=1.2)

plt.tight_layout()
plt.show()
```

## Gradient of a Scalar Field

### Laplacian Operator

**Gradient Computation and Visualization**

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a scalar field on a grid
x = np.linspace(-2, 2, 30)
y = np.linspace(-2, 2, 30)
X, Y = np.meshgrid(x, y)

# Scalar field: f(x,y) = exp(-(x**2 + y**2))
f = np.exp(-(X**2 + Y**2))

# Compute gradient using numerical differentiation
# With meshgrid 'xy' indexing, axis 0 is y and axis 1 is x
dx = x[1] - x[0]
dy = y[1] - y[0]
grad_y, grad_x = np.gradient(f, dy, dx)

# Plot contours and gradient vectors
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Subplot 1: Contour plot with gradient vectors
ax = axes[0]
contour = ax.contour(X, Y, f, levels=10, colors='black', alpha=0.3)
ax.clabel(contour, inline=True, fontsize=8)
# Downsample for cleaner visualization
skip = 2
ax.quiver(X[::skip, ::skip], Y[::skip, ::skip],
          grad_x[::skip, ::skip], grad_y[::skip, ::skip],
          color='red', alpha=0.7)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Level Curves and Gradient Vectors')
ax.grid(True, alpha=0.3)

# Subplot 2: 3D surface
ax = fig.add_subplot(122, projection='3d')
ax.plot_surface(X, Y, f, cmap='viridis', alpha=0.8)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x,y)')
ax.set_title('Scalar Field f(x,y) = exp(-(x**2 + y**2))')

plt.tight_layout()
plt.show()

# Compute analytical gradient for verification
# nablaf = -2(x*i + y*j)*exp(-(x**2 + y**2))
grad_x_analytical = -2*X*np.exp(-(X**2 + Y**2))
grad_y_analytical = -2*Y*np.exp(-(X**2 + Y**2))

# Check magnitude of gradient
grad_magnitude = np.sqrt(grad_x**2 + grad_y**2)
print(f"Maximum gradient magnitude: {grad_magnitude.max():.4f}")
print(f"Gradient is largest near the origin and decreases away from it.")
```

## Divergence of a Vector Field

### Flux Interpretation

**Divergence Computation and Visualization**

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a vector field on a grid
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)

# Vector field 1: Radial expansion (positive divergence)
# F = (x*i + y*j) - represents outward flow from origin
F1_x = X
F1_y = Y
div_F1 = 1 + 1  # div(F1) = partialx/partialx + partialy/partialy = 2 everywhere

# Vector field 2: Rotation (zero divergence)
# F = (-y*i + x*j) - rotational field
F2_x = -Y
F2_y = X
div_F2 = 0  # div(F2) = partial(-y)/partialx + partialx/partialy = 0

# Vector field 3: Convergence (negative divergence)
# F = (-x*i - y*j) - represents inward flow toward origin
F3_x = -X
F3_y = -Y
div_F3 = -1 + (-1)  # div(F3) = -1 - 1 = -2 everywhere

# Create visualization
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Radial expansion
ax = axes[0, 0]
ax.quiver(X, Y, F1_x, F1_y, alpha=0.6)
ax.set_title('Radial Expansion: div F = 2 (Source)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Plot 2: Rotation
ax = axes[0, 1]
ax.quiver(X, Y, F2_x, F2_y, alpha=0.6, color='green')
ax.set_title('Rotation: div F = 0 (No source/sink)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Plot 3: Convergence
ax = axes[1, 0]
ax.quiver(X, Y, F3_x, F3_y, alpha=0.6, color='red')
ax.set_title('Convergence: div F = -2 (Sink)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Plot 4: Complex vector field with varying divergence
# F = ((1 - x**2)*i + (-y)*j)
F4_x = 1 - X**2
F4_y = -Y
# Compute divergence numerically
# With meshgrid 'xy' indexing, x varies along axis 1 and y along axis 0
dx = x[1] - x[0]
dy = y[1] - y[0]
dF4x_dx = np.gradient(F4_x, dx, axis=1)
dF4y_dy = np.gradient(F4_y, dy, axis=0)
div_F4 = dF4x_dx + dF4y_dy

ax = axes[1, 1]
divergence_plot = ax.contourf(X, Y, div_F4, levels=15, cmap='RdBu_r')
ax.quiver(X, Y, F4_x, F4_y, alpha=0.4, color='black', scale=40)
plt.colorbar(divergence_plot, ax=ax, label='div F')
ax.set_title('Complex Field with Varying Divergence')
ax.set_xlabel('x')
ax.set_ylabel('y')

plt.tight_layout()
plt.show()
```

## Curl of a Vector Field

### Finding Potential Functions

**Curl Computation and Conservative Field Analysis**

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a vector field on a grid
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)

# Vector field 1: Rotational (non-zero curl)
# F = (-y*i + x*j) - rotation around z-axis
F1_x = -Y
F1_y = X
# Curl: (partialFy/partialx - partialFx/partialy)_z = (1 - (-1)) = 2

# Vector field 2: Conservative/potential field (zero curl)
# F = (2x*i + 2y*j) = grad(x**2 + y**2)
F2_x = 2*X
F2_y = 2*Y
# Curl: (partial(2y)/partialx - partial(2x)/partialy)_z = (0 - 0) = 0

# Vector field 3: Another rotational field
# F = (y*i + 0*j) - shear flow
F3_x = Y
F3_y = np.zeros_like(Y)
# Curl: (partial(0)/partialx - partial(y)/partialy)_z = (0 - 1) = -1

# Create visualization
fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# Function to compute 2D curl (z-component only)
def compute_curl_z(Fx, Fy, dx, dy):
    """Compute z-component of curl: partialFy/partialx - partialFx/partialy.
    With meshgrid 'xy' indexing, x varies along axis 1 and y along axis 0."""
    dFy_dx = np.gradient(Fy, dx, axis=1)
    dFx_dy = np.gradient(Fx, dy, axis=0)
    return dFy_dx - dFx_dy

dx = x[1] - x[0]
dy = y[1] - y[0]
curl_z1 = compute_curl_z(F1_x, F1_y, dx, dy)
curl_z2 = compute_curl_z(F2_x, F2_y, dx, dy)
curl_z3 = compute_curl_z(F3_x, F3_y, dx, dy)

# Plot 1: Rotational field
ax = axes[0, 0]
ax.quiver(X, Y, F1_x, F1_y, alpha=0.6, color='blue')
curl_plot1 = ax.contourf(X, Y, curl_z1, levels=10, cmap='RdBu_r', alpha=0.3)
plt.colorbar(curl_plot1, ax=ax, label='curl z')
ax.set_title('Rotational Field: nablaxF != 0\nF = (-y, x, 0)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Plot 2: Conservative field
ax = axes[0, 1]
ax.quiver(X, Y, F2_x, F2_y, alpha=0.6, color='green')
curl_plot2 = ax.contourf(X, Y, curl_z2, levels=10, cmap='RdBu_r', alpha=0.3)
plt.colorbar(curl_plot2, ax=ax, label='curl z')
ax.set_title('Conservative Field: nablaxF = 0\nF = (2x, 2y, 0) = nabla(x**2 + y**2)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Plot 3: Shear field
ax = axes[1, 0]
ax.quiver(X, Y, F3_x, F3_y, alpha=0.6, color='red')
curl_plot3 = ax.contourf(X, Y, curl_z3, levels=10, cmap='RdBu_r', alpha=0.3)
plt.colorbar(curl_plot3, ax=ax, label='curl z')
ax.set_title('Shear Field: nablaxF != 0\nF = (y, 0, 0)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Plot 4: Summary visualization of curl interpretation
ax = axes[1, 1]
# More complex field
F_x = -Y / (X**2 + Y**2 + 0.1)
F_y = X / (X**2 + Y**2 + 0.1)
curl_z_complex = compute_curl_z(F_x, F_y, dx, dy)

ax.quiver(X, Y, F_x, F_y, alpha=0.6, color='black')
curl_plot4 = ax.contourf(X, Y, curl_z_complex, levels=15, cmap='RdBu_r', alpha=0.4)
plt.colorbar(curl_plot4, ax=ax, label='curl z')
ax.set_title('Complex Vortex Field')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Verify properties
print("Curl Property Verification:")
print(f"Rotational field curl (z-component) ranges: [{curl_z1.min():.2f}, {curl_z1.max():.2f}]")
print(f"Conservative field curl (z-component) ranges: [{curl_z2.min():.6f}, {curl_z2.max():.6f}] ~= 0")
print(f"Shear field curl (z-component) ranges: [{curl_z3.min():.2f}, {curl_z3.max():.2f}]")
```

## Line Integrals

### Applications of Green's Theorem

**Line Integrals and Green's Theorem Computation**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

# Example 1: Line integral along a parametric curve
def line_integral_example1():
    """
    Compute integral of f = x*y along C: the quarter circle x^2 + y^2 = 1
    from (1,0) to (0,1)
    """
    # Parametrization: r(t) = (cos(t), sin(t)), t in [0, pi/2]
    def f(t):
        x = np.cos(t)
        y = np.sin(t)
        r_prime_mag = 1  # |r'(t)| = 1
        return x * y * r_prime_mag

    integral, _ = quad(f, 0, np.pi/2)
    print(f"Line integral of f = x*y along quarter circle: {integral:.4f}")

# Example 2: Work integral with vector field
def line_integral_example2():
    """
    Compute work done by F = (y, x, 0) along C: unit circle from t=0 to t=2pi
    """
    def integrand(t):
        # Position and tangent
        x, y = np.cos(t), np.sin(t)
        dx_dt, dy_dt = -np.sin(t), np.cos(t)

        # Force field F = (y, x, 0)
        Fx, Fy = y, x

        # F * dr = F * r'(t) dt
        return Fx * dx_dt + Fy * dy_dt

    work, _ = quad(integrand, 0, 2*np.pi)
    print(f"Work done by F = (y, x, 0) around unit circle: {work:.6f}")

# Example 3: Green's theorem - compute area of ellipse
def greens_theorem_area():
    """
    Use Green's theorem to compute area of ellipse: x**2/4 + y**2/9 = 1
    """
    a, b = 2, 3  # Semi-axes

    # Parametrization: r(t) = (a*cos(t), b*sin(t)), t in [0, 2pi]
    def integrand(t):
        x = a * np.cos(t)
        y = b * np.sin(t)
        dx_dt = -a * np.sin(t)
        dy_dt = b * np.cos(t)

        # A = (1/2) contour_integral (x dy - y dx)
        return 0.5 * (x * dy_dt - y * dx_dt)

    area, _ = quad(integrand, 0, 2*np.pi)
    analytical_area = np.pi * a * b

    print(f"\nGreen's Theorem - Ellipse Area:")
    print(f"Computed area: {area:.4f}")
    print(f"Analytical area (pi*a*b): {analytical_area:.4f}")
    print(f"Error: {abs(area - analytical_area):.6f}")

# Example 4: Verify Green's theorem
def verify_greens_theorem():
    """
    Verify Green's theorem: contour_integral (P dx + Q dy) = double_integral (partialQ/partialx - partialP/partialy) dA

    Use: P = -y, Q = x (which is nablaxF in 2D gives curl_z = 2)
    Region: unit circle
    """
    # Line integral around unit circle
    def line_integral(t):
        x, y = np.cos(t), np.sin(t)
        dx_dt, dy_dt = -np.sin(t), np.cos(t)
        P, Q = -y, x
        return P * dx_dt + Q * dy_dt

    line_int, _ = quad(line_integral, 0, 2*np.pi)

    # Double integral over disk: partialQ/partialx - partialP/partialy = 1 - (-1) = 2
    double_int = 2 * np.pi * 1  # 2 * Area(unit circle)

    print(f"\nGreen's Theorem Verification (Unit Circle):")
    print(f"Line integral contour_integral (-y dx + x dy): {line_int:.4f}")
    print(f"Double integral double_integral 2 dA: {double_int:.4f}")
    print(f"Match: {np.isclose(line_int, double_int)}")

# Run examples
line_integral_example1()
line_integral_example2()
greens_theorem_area()
verify_greens_theorem()
```

## Surface Integrals

### Flux Integral

**Surface Integrals and Flux Computation**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate

# Example 1: Surface area of a sphere
def sphere_surface_area():
    """
    Compute surface area of unit sphere using surface integral
    """
    # Spherical parametrization
    def integrand(phi, theta):
        # r(theta,phi) = [sin(phi)cos(theta), sin(phi)sin(theta), cos(phi)]
        # |dr/dtheta x dr/dphi| = sin(phi)
        return np.sin(phi)

    area, _ = integrate.dblquad(integrand, 0, 2*np.pi, 0, np.pi)
    analytical = 4 * np.pi

    print(f"Surface Area of Unit Sphere:")
    print(f"Computed: {area:.4f}")
    print(f"Analytical (4pi): {analytical:.4f}")

# Example 2: Flux of radial field through sphere
def flux_radial_field():
    """
    Compute flux of F = (x, y, z) through unit sphere
    """
    # On unit sphere: F*n = (x, y, z)*(x, y, z) = x**2 + y**2 + z**2 = 1
    # So flux = double_integral 1 dS = surface area = 4pi

    flux_analytical = 4 * np.pi
    print(f"\nFlux of Radial Field F = (x,y,z) through Unit Sphere:")
    print(f"Analytical flux: {flux_analytical:.4f}")

# Example 3: Flux through paraboloid
def flux_through_paraboloid():
    """
    Compute flux of F = (0, 0, 1) (constant upward) through z = x**2 + y**2 (below z=1)
    """
    # Parametrization: r(r, theta) = [r*cos(theta), r*sin(theta), r**2]
    # with 0 <= r <= 1, 0 <= theta < 2pi

    # partialr/partialr = [cos(theta), sin(theta), 2r]
    # partialr/partialtheta = [-r*sin(theta), r*cos(theta), 0]
    # Cross product: partialr/partialr x partialr/partialtheta = [-2r**2cos(theta), -2r**2sin(theta), r]

    # F*(partialr/partialr x partialr/partialtheta) = [0, 0, 1]*[-2r**2cos(theta), -2r**2sin(theta), r] = r

    def integrand(theta, r):
        return r  # r from dot product

    flux, _ = integrate.dblquad(integrand, 0, 1, 0, 2*np.pi)
    print(f"\nFlux of F = (0,0,1) through Paraboloid z = x**2 + y**2:")
    print(f"Computed flux: {flux:.4f}")
    print(f"Analytical (pi): {np.pi:.4f}")

# Run examples
sphere_surface_area()
flux_radial_field()
flux_through_paraboloid()
```

## Integral Theorems

### Applications of Integral Theorems

**Stokes' and Divergence Theorems Verification**

```python
import numpy as np
from scipy import integrate

# Example 1: Stokes' Theorem verification
def stokes_theorem_verification():
    """
    Verify Stokes' theorem with F = (-y, x, 0) on unit disk in xy-plane
    """
    # Line integral around unit circle
    def line_integral(t):
        x, y = np.cos(t), np.sin(t)
        dx_dt, dy_dt = -np.sin(t), np.cos(t)
        Fx, Fy = -y, x
        return Fx * dx_dt + Fy * dy_dt

    line_int, _ = integrate.quad(line_integral, 0, 2*np.pi)

    # Surface integral of curl
    # nablaxF = [0, 0, partial(x)/partialx - partial(-y)/partialy] = [0, 0, 2]
    # double_integral (curl F)*n dS = double_integral 2 dS = 2 * pi (area of unit disk)
    surface_int = 2 * np.pi

    print("Stokes' Theorem Verification (Unit Disk):")
    print(f"Line integral contour_integral (-y dx + x dy): {line_int:.4f}")
    print(f"Surface integral double_integral (nablaxF)*n dS: {surface_int:.4f}")
    print(f"Match: {np.isclose(line_int, surface_int)}\n")

# Example 2: Divergence Theorem verification
def divergence_theorem_verification():
    """
    Verify divergence theorem with F = (x, y, z) on unit ball
    """
    # Surface integral: flux of F through sphere
    # F*n = (x, y, z)*(x, y, z)/1 = 1 on unit sphere
    # double_integral F*n dS = surface area = 4pi
    surface_int = 4 * np.pi

    # Volume integral: nabla*F = dx/dx + dy/dy + dz/dz = 3
    # Volume integral of 3 dV = 3 * (4pi/3) = 4pi
    volume_int = 4 * np.pi

    print("Divergence Theorem Verification (Unit Ball):")
    print(f"Surface integral of F*n dS: {surface_int:.4f}")
    print(f"Volume integral of nabla*F dV: {volume_int:.4f}")
    print(f"Match: {np.isclose(surface_int, volume_int)}\n")

# Example 3: Physical interpretation - flux and divergence
def flux_divergence_example():
    """
    Compute flux for F = (0, 0, z) through cylinder
    """
    # Cylinder: x**2 + y**2 = 1, 0 <= z <= 1

    # Side surface (cylindrical): n = (cos(theta), sin(theta), 0)
    # F*n = 0, so side contributes 0

    # Top cap (z = 1): n = (0, 0, 1)
    # F*n = 1, flux = double_integral 1 dA = pi

    # Bottom cap (z = 0): n = (0, 0, -1)
    # F*n = 0, flux = 0

    # Total flux = pi

    # Via divergence theorem:
    # nabla*F = d(0)/dx + d(0)/dy + d(z)/dz = 1
    # Volume integral of 1 dV = Volume of cylinder = pi * 1 = pi

    surface_flux = np.pi
    volume_integral = np.pi

    print("Flux Through Cylinder (Divergence Theorem):")
    print(f"Surface integral (direct): {surface_flux:.4f}")
    print(f"Volume integral of nabla*F dV: {volume_integral:.4f}")
    print(f"Match: {np.isclose(surface_flux, volume_integral)}\n")

# Run verifications
stokes_theorem_verification()
divergence_theorem_verification()
flux_divergence_example()
```

## Curvilinear Coordinates

### Applications to PDEs

**Coordinate Transformation and Differential Operators**

```python
import numpy as np
import matplotlib.pyplot as plt

# Example 1: Coordinate transformations
def coordinate_transformations():
    """
    Demonstrate conversions between coordinate systems
    """
    # Cartesian point
    x, y, z = 1.0, 1.0, 2.0

    # Convert to cylindrical
    r_cyl = np.sqrt(x**2 + y**2)
    theta = np.arctan2(y, x)
    z_cyl = z

    print("Coordinate Transformations:")
    print(f"Cartesian: (x, y, z) = ({x:.2f}, {y:.2f}, {z:.2f})")
    print(f"Cylindrical: (r, theta, z) = ({r_cyl:.4f}, {theta:.4f} rad, {z_cyl:.2f})")

    # Convert to spherical
    r_sph = np.sqrt(x**2 + y**2 + z**2)
    theta_sph = np.arctan2(y, x)
    phi = np.arccos(z / r_sph)

    print(f"Spherical: (r, theta, phi) = ({r_sph:.4f}, {theta_sph:.4f} rad, {phi:.4f} rad)")

    # Verify inverse transformation
    x_back = r_sph * np.sin(phi) * np.cos(theta_sph)
    y_back = r_sph * np.sin(phi) * np.sin(theta_sph)
    z_back = r_sph * np.cos(phi)

    print(f"Back to Cartesian: ({x_back:.4f}, {y_back:.4f}, {z_back:.4f})\n")

# Example 2: Laplacian in cylindrical coordinates
def laplacian_cylindrical():
    """
    Compute Laplacian of f(r) = r**2 in cylindrical coordinates
    """
    # f(r) = r**2
    # nabla**2f = (1/r) d/dr(r df/dr)
    # df/dr = 2r
    # r df/dr = 2r**2
    # d/dr(2r**2) = 4r
    # nabla**2f = (1/r) * 4r = 4

    r_vals = np.linspace(0.1, 2, 20)
    laplacian_values = 4 * np.ones_like(r_vals)  # Constant = 4

    # Verify numerically
    dr = 0.01
    def f(r):
        return r**2

    laplacian_numerical = []
    for r in r_vals:
        if r < dr:
            continue
        # (1/r) * d/dr(r * df/dr)
        f_plus = f(r + dr)
        f_center = f(r)
        f_minus = f(r - dr)

        df_dr = (f_plus - f_minus) / (2*dr)
        d2f = (f_plus - 2*f_center + f_minus) / (dr**2)

        laplacian_num = (1/r) * (d2f * r + df_dr)  # Using product rule
        laplacian_numerical.append(laplacian_num)

    print("Laplacian in Cylindrical Coordinates (f = r**2):")
    print(f"Analytical nabla**2f = 4")
    print(f"Numerical (sample) nabla**2f at r=1.0: {laplacian_numerical[int(len(laplacian_numerical)/2)]:.4f}\n")

# Example 3: Spherical harmonics
def spherical_harmonics():
    """
    Visualize spherical coordinate basis and Laplacian
    """
    theta = np.linspace(0, 2*np.pi, 50)
    phi = np.linspace(0, np.pi, 30)
    THETA, PHI = np.meshgrid(theta, phi)

    # Convert to Cartesian for visualization
    r_sph = 1  # Unit sphere
    X = r_sph * np.sin(PHI) * np.cos(THETA)
    Y = r_sph * np.sin(PHI) * np.sin(THETA)
    Z = r_sph * np.cos(PHI)

    # A function on the sphere: f = cos(phi)
    f = np.cos(PHI)

    fig = plt.figure(figsize=(10, 4))

    # Plot 1: 3D surface colored by function value
    ax = fig.add_subplot(121, projection='3d')
    ax.plot_surface(X, Y, Z, facecolors=plt.cm.RdBu_r(
        (f - f.min()) / (f.max() - f.min())), shade=False)
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title('Function on Sphere: f = cos(phi)')

    # Plot 2: Function value vs polar angle
    ax = fig.add_subplot(122)
    phi_1d = np.linspace(0, np.pi, 100)
    f_1d = np.cos(phi_1d)
    ax.plot(phi_1d, f_1d, 'b-', linewidth=2)
    ax.set_xlabel('Polar angle phi (radians)')
    ax.set_ylabel('f = cos(phi)')
    ax.set_title('Function Profile')
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

# Run examples
coordinate_transformations()
laplacian_cylindrical()
spherical_harmonics()
```

## Applications in Engineering

### Heat Transfer

**Numerical Solutions: Fluid Flow and Heat Transfer**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import sparse
from scipy.sparse.linalg import spsolve

# Example 1: 2D Poisson equation for streamfunction (inviscid flow)
def streamfunction_flow():
    """
    Solve Laplacian of psi = omega (vorticity) for 2D incompressible flow
    Using finite differences on a grid
    """
    nx, ny = 30, 30
    x = np.linspace(0, 1, nx)
    y = np.linspace(0, 1, ny)
    dx = x[1] - x[0]
    dy = y[1] - y[0]

    # Create vorticity field (example: circular vortex)
    X, Y = np.meshgrid(x, y)
    omega = -2 * np.exp(-((X - 0.5)**2 + (Y - 0.5)**2) / 0.01)  # Gaussian vortex

    # Solve Poisson equation: Laplacian(psi) = omega using finite differences
    n = nx * ny
    A = sparse.lil_matrix((n, n))
    b = np.zeros(n)

    for i in range(1, ny-1):
        for j in range(1, nx-1):
            idx = i*nx + j
            # 5-point Laplacian stencil at (i,j)
            A[idx, idx] = -2.0/(dx*dx) - 2.0/(dy*dy)
            A[idx, (i-1)*nx + j] = 1.0/(dy*dy)
            A[idx, (i+1)*nx + j] = 1.0/(dy*dy)
            A[idx, i*nx + (j-1)] = 1.0/(dx*dx)
            A[idx, i*nx + (j+1)] = 1.0/(dx*dx)
            b[idx] = omega[i, j]

    # Boundary conditions (psi = 0 on boundaries)
    for j in range(nx):
        A[0*nx + j, 0*nx + j] = 1.0
        A[(ny-1)*nx + j, (ny-1)*nx + j] = 1.0
        b[0*nx + j] = 0.0
        b[(ny-1)*nx + j] = 0.0

    for i in range(ny):
        A[i*nx + 0, i*nx + 0] = 1.0
        A[i*nx + (nx-1), i*nx + (nx-1)] = 1.0
        b[i*nx + 0] = 0.0
        b[i*nx + (nx-1)] = 0.0

    # Solve system
    A = A.tocsr()
    psi = spsolve(A, b)
    psi = psi.reshape((ny, nx))

    # Compute velocity components from streamfunction
    v_x = np.zeros_like(psi)
    v_y = np.zeros_like(psi)
    for i in range(1, ny-1):
        for j in range(1, nx-1):
            v_x[i, j] = (psi[i+1, j] - psi[i-1, j]) / (2*dy)
            v_y[i, j] = -(psi[i, j+1] - psi[i, j-1]) / (2*dx)

    # Visualization
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    ax = axes[0]
    contour = ax.contourf(X, Y, psi, levels=20, cmap='viridis')
    ax.contour(X, Y, psi, levels=10, colors='black', alpha=0.3, linewidths=0.5)
    plt.colorbar(contour, ax=ax, label='Streamfunction psi')
    ax.set_title('Streamfunction (psi) for Vortex Flow')
    ax.set_xlabel('x')
    ax.set_ylabel('y')

    ax = axes[1]
    skip = 2
    ax.quiver(X[::skip, ::skip], Y[::skip, ::skip],
              v_x[::skip, ::skip], v_y[::skip, ::skip], alpha=0.7)
    ax.set_title('Velocity Field (from Streamfunction)')
    ax.set_xlabel('x')
    ax.set_ylabel('y')

    plt.tight_layout()
    plt.show()

# Example 2: Heat diffusion in 1D
def heat_diffusion_1d():
    """
    Solve 1D heat equation: partialT/partialt = alpha partial**2T/partialx**2
    Using finite difference method (FTCS: Forward Time Centered Space)
    """
    # Parameters
    L = 1.0  # Domain length
    T_final = 0.1
    alpha = 0.01  # Thermal diffusivity

    # Discretization
    nx, nt = 50, 100
    dx = L / nx
    dt = T_final / nt
    r = alpha * dt / (dx**2)  # Courant number

    print(f"\n1D Heat Diffusion:")
    print(f"Courant number r = {r:.4f}")
    if r > 0.5:
        print("Warning: r > 0.5, method may be unstable")

    # Initial condition: T(x,0) = sin(pix)
    x = np.linspace(0, L, nx)
    T = np.sin(np.pi * x)
    T_history = [T.copy()]

    # Time stepping
    for n in range(nt):
        T_new = T.copy()
        for i in range(1, nx-1):
            T_new[i] = T[i] + r * (T[i+1] - 2*T[i] + T[i-1])

        # Boundary conditions: T(0,t) = T(1,t) = 0
        T_new[0] = 0
        T_new[-1] = 0

        T = T_new
        T_history.append(T.copy())

    T_history = np.array(T_history)

    # Analytical solution: T(x,t) = exp(-pi**2alphat) * sin(pix)
    t_analytical = np.linspace(0, T_final, nt+1)
    T_analytical = np.exp(-np.pi**2 * alpha * t_analytical[-1]) * np.sin(np.pi * x)

    # Visualization
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    # Time evolution
    ax = axes[0]
    times = [0, nt//4, nt//2, nt]
    colors = ['blue', 'green', 'orange', 'red']
    for n, color in zip(times, colors):
        ax.plot(x, T_history[n], color=color, label=f't = {n*dt:.4f}')
    ax.set_xlabel('Position x')
    ax.set_ylabel('Temperature T')
    ax.set_title('1D Heat Diffusion: Temperature Evolution')
    ax.legend()
    ax.grid(True, alpha=0.3)

    # Heatmap of T(x,t)
    ax = axes[1]
    time = np.linspace(0, T_final, nt+1)
    contour = ax.contourf(x, time, T_history, levels=20, cmap='hot')
    plt.colorbar(contour, ax=ax, label='Temperature T')
    ax.set_xlabel('Position x')
    ax.set_ylabel('Time t')
    ax.set_title('Temperature Profile Over Time')

    plt.tight_layout()
    plt.show()

# Run examples
streamfunction_flow()
heat_diffusion_1d()
```
