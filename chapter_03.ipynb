## Homogeneous Linear ODEs

### Symbolic Solutions with SymPy

```python
import sympy as sp

# Define the symbolic variable
x = sp.Symbol('x')

# Define two candidate solutions
y1 = sp.exp(x)
y2 = sp.exp(-x)

# Compute their derivatives
y1_prime = sp.diff(y1, x)
y2_prime = sp.diff(y2, x)

# Compute the Wronskian determinant
W = sp.Matrix([[y1, y2], [y1_prime, y2_prime]]).det()

# Display and simplify the result
print("Wronskian W(e^x, e^-x) =", W)
print("Simplified:", sp.simplify(W))

# Verify that both functions satisfy the ODE y'' - y = 0
diff_eq = lambda y: sp.diff(y, x, 2) - y

# Check if y1 satisfies the ODE
result1 = diff_eq(y1)
print(f"{y1} satisfies the ODE:", sp.simplify(result1) == 0)

# Check if y2 satisfies the ODE
result2 = diff_eq(y2)
print(f"{y2} satisfies the ODE:", sp.simplify(result2) == 0)
```

## Homogeneous Linear ODEs with Constant Coefficients

### Symbolic and Numerical Solutions with Python

```python
import sympy as sp

# Define symbolic variables
x = sp.Symbol('x')
y = sp.Function('y')(x)

# Case 1: Distinct real roots (y'' + y' - 2y = 0)
ode1 = sp.Eq(sp.diff(y, x, 2) + sp.diff(y, x) - 2*y, 0)
solution1 = sp.dsolve(ode1, y)
print('ODE with distinct real roots solution:', solution1)

# Case 2: Repeated real root (y'' + 4y' + 4y = 0)
ode2 = sp.Eq(sp.diff(y, x, 2) + 4*sp.diff(y, x) + 4*y, 0)
solution2 = sp.dsolve(ode2, y)
print('ODE with repeated real root solution:', solution2)

# Case 3: Complex conjugate roots (y'' + 4y = 0)
ode3 = sp.Eq(sp.diff(y, x, 2) + 4*y, 0)
solution3 = sp.dsolve(ode3, y)
print('ODE with complex conjugate roots solution:', solution3)
```

```python
ODE with distinct real roots solution: Eq(y(x), C1*exp(-2*x) + C2*exp(x))
ODE with repeated real root solution: Eq(y(x), (C1 + C2*x)*exp(-2*x))
ODE with complex conjugate roots solution: Eq(y(x), C1*sin(2*x) + C2*cos(2*x))
```

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Define the ODE system as first-order equations
def ode_system(t, Y, a, b, c):
    y, dy_dt = Y
    d2y_dt2 = -(b*dy_dt + c*y)/a
    return [dy_dt, d2y_dt2]

# Set up parameters for y'' + 4y' + 4y = 0
a, b, c = 1, 4, 4
initial_conditions = [1, 0]  # y(0) = 1, y'(0) = 0

# Solve numerically
t_span = (0, 5)
t_eval = np.linspace(0, 5, 1000)
solution = solve_ivp(
    lambda t, Y: ode_system(t, Y, a, b, c),
    t_span, initial_conditions, t_eval=t_eval
)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(solution.t, solution.y[0], 'b-', linewidth=2)
plt.grid(True)
plt.xlabel('t', fontsize=14)
plt.ylabel('y(t)', fontsize=14)
plt.title('Solution to y\'\' + 4y\' + 4y = 0 with y(0)=1, y\'(0)=0', fontsize=16)
```

## Free Oscillations of a Mass-Spring System

### Computational Analysis

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Symbolic solution
t = sp.Symbol('t', real=True)
m, k = sp.symbols('m k', positive=True)
omega = sp.sqrt(k/m)
A, B = sp.symbols('A B')

# General solution
x = A*sp.cos(omega*t) + B*sp.sin(omega*t)

# Convert to phase-amplitude form
C = sp.sqrt(A**2 + B**2)
phi = sp.atan2(A, B)
x_phase = C*sp.sin(omega*t + phi)

print("Equivalence of forms:", sp.simplify(x - x_phase))

# Numerical simulation for specific parameters
def spring_mass_system(t, y, m, k):
    """
    First-order ODE system for the spring-mass equation
    y[0] = position, y[1] = velocity
    """
    return [y[1], -k/m * y[0]]

# Parameters
m_val = 2.0  # kg
k_val = 8.0  # N/m
omega_val = np.sqrt(k_val/m_val)
print(f"Natural frequency: {omega_val} rad/s")

# Initial conditions: x(0) = 0.5, x'(0) = 0
y0 = [0.5, 0]

# Time span
t_span = (0, 10)
t_eval = np.linspace(0, 10, 1000)

# Solve the ODE system
solution = solve_ivp(
    lambda t, y: spring_mass_system(t, y, m_val, k_val),
    t_span, y0, t_eval=t_eval, method='RK45'
)

# Compare with analytical solution
t_points = solution.t
analytical = 0.5 * np.cos(omega_val * t_points)

# Plot both solutions
plt.figure(figsize=(12, 6))
plt.plot(t_points, solution.y[0], 'b-', label='Numerical')
plt.plot(t_points, analytical, 'r--', label='Analytical')
plt.grid(True)
plt.xlabel('Time (s)', fontsize=14)
plt.ylabel('Position (m)', fontsize=14)
plt.title('Mass-Spring System: Free Oscillation', fontsize=16)
plt.legend()
```

### Comprehensive Example: Mass–Spring Free Oscillation

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Symbolic solution
t = sp.Symbol('t', real=True)
m, k = sp.symbols('m k', positive=True)
x = sp.Function('x')(t)

# Define the ODE
ode = sp.Eq(m*sp.diff(x, t, 2) + k*x, 0)
print("ODE:", ode)

# Substitute values
ode_subst = ode.subs({m: 2, k: 8})
print("ODE with values:", ode_subst)

# Solve the ODE
general_sol = sp.dsolve(ode_subst, x)
print("General solution:", general_sol)

# Apply initial conditions
constants = sp.symbols('C1 C2')
x_sol = general_sol.rhs
x_prime = sp.diff(x_sol, t)

# Initial conditions: x(0) = 0.5, x'(0) = 0
eq1 = sp.Eq(x_sol.subs(t, 0), 0.5)
eq2 = sp.Eq(x_prime.subs(t, 0), 0)

# Solve for constants
const_values = sp.solve((eq1, eq2), constants)
print("Constants:", const_values)

# Substitute constants
particular_sol = x_sol.subs(const_values)
print("Particular solution:", particular_sol)

# Create function for numerical evaluation
x_func = sp.lambdify(t, particular_sol, 'numpy')

# Generate time points and calculate positions
t_vals = np.linspace(0, 5, 1000)
x_vals = x_func(t_vals)

# Plot the solution
plt.figure(figsize=(10, 6))
plt.plot(t_vals, x_vals, 'b-', linewidth=2)
plt.grid(True)
plt.xlabel('Time (s)', fontsize=14)
plt.ylabel('Position (m)', fontsize=14)
plt.title('Mass-Spring Motion: x(t) = 0.5cos(2t)', fontsize=16)
plt.savefig('mass_spring_oscillation.png')
```

## Damped Harmonic Motion

### Computational Analysis of Damped Systems

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
import sympy as sp

# Define the damped harmonic oscillator system
def damped_oscillator(t, y, zeta, omega0):
    """
    First-order ODE system for the damped oscillator equation
    y[0] = position, y[1] = velocity
    """
    return [
        y[1],
        -2*zeta*omega0*y[1] - omega0**2*y[0]
    ]

# Parameters
omega0 = 4.0  # Natural frequency (rad/s)
initial_displacement = 1.0
initial_velocity = 0.0
t_span = (0, 5)
t_eval = np.linspace(0, 5, 1000)

# Initial conditions
y0 = [initial_displacement, initial_velocity]

# Create a figure for comparison
plt.figure(figsize=(12, 8))

# Simulate for different damping ratios
damping_ratios = {
    'Underdamped': 0.2,
    'Critically damped': 1.0,
    'Overdamped': 1.5
}

for label, zeta in damping_ratios.items():
    # Solve the ODE system
    solution = solve_ivp(
        lambda t, y: damped_oscillator(t, y, zeta, omega0),
        t_span, y0, t_eval=t_eval, method='RK45'
    )
    
    # Plot the solution
    plt.plot(solution.t, solution.y[0], label=fr'{label} ($\zeta = {zeta}$)')

# Add mathematical formulas for each case
plt.grid(True)
plt.xlabel('Time (s)', fontsize=14)
plt.ylabel('Position (m)', fontsize=14)
plt.title('Damped Harmonic Oscillator: Effect of Damping Ratio', fontsize=16)
plt.legend()
plt.savefig('damping_comparison.png', dpi=300, bbox_inches='tight')

# Symbolic solution for underdamped case. Use new names for the
# symbols so they do not shadow the floats omega0 and zeta above.
t = sp.Symbol('t', real=True)
zeta_s, omega0_s = sp.symbols('zeta omega0', real=True, positive=True)

# Assume 0 < zeta < 1 (underdamped)
omega_d = omega0_s * sp.sqrt(1 - zeta_s**2)
A, B = sp.symbols('A B', real=True)

# Underdamped solution
x_underdamped = sp.exp(-zeta_s*omega0_s*t) * (A*sp.cos(omega_d*t)
                                              + B*sp.sin(omega_d*t))
print("Symbolic underdamped solution:", x_underdamped)

# Verify that it satisfies the ODE
ode_lhs = (sp.diff(x_underdamped, t, 2)
           + 2*zeta_s*omega0_s*sp.diff(x_underdamped, t)
           + omega0_s**2*x_underdamped)
print("Residual (should be 0):", sp.simplify(ode_lhs))
```

## Euler-Cauchy Equations

### Computational Implementation

```python
import sympy as sp

# Define symbolic variables
x = sp.Symbol('x', positive=True)  # Restrict to positive x for logarithm
y = sp.Function('y')(x)

# Define an Euler-Cauchy equation
a, b, c = sp.symbols('a b c')
euler_cauchy = a*x**2*y.diff(x, 2) + b*x*y.diff(x) + c*y

# Example equation: x^2*y'' - 3x*y' + 4y = 0
eq = sp.Eq(x**2*y.diff(x, 2) - 3*x*y.diff(x) + 4*y, 0)
print("Euler-Cauchy equation:", eq)

# Solve the equation
solution = sp.dsolve(eq, y)
print("Solution:", solution)

# Manual substitution approach
# Define t = ln(x) => x = e^t
t = sp.Symbol('t', real=True)
x_expr = sp.exp(t)

# Define the derivatives in terms of t
dy_dx = sp.Function('Y')(t).diff(t) / x_expr
d2y_dx2 = (sp.Function('Y')(t).diff(t, 2) - sp.Function('Y')(t).diff(t)) / x_expr**2

# Substitute into the original equation
transformed_eq = a*(x_expr**2)*d2y_dx2 + b*x_expr*dy_dx + c*sp.Function('Y')(t)
print("Transformed equation (manual):", transformed_eq.simplify())

# This should result in a constant coefficient equation:
# a*Y''(t) + (b-a)*Y'(t) + c*Y(t) = 0
```

## Green's Functions for ODEs

### Python Implementation with SymPy

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Define the symbolic variable
x, xi, lam = sp.symbols('x xi lambda', real=True, positive=True)

# For the problem: d^2u/dx^2 + lambda*u = f(x)
# with u(0) = 0, u(1) = 0
# Boundary conditions require u = 0 at x = 0 and x = 1

# Homogeneous solutions satisfying boundary conditions
# u_1(x) = sin(sqrt(lambda)*x)  (satisfies u(0)=0)
# u_2(x) = sin(sqrt(lambda)*(1-x))  (satisfies u(1)=0)

lam_val = 1
sqrt_lam = np.sqrt(lam_val)

# Define symbolic functions
u1 = sp.sin(sqrt_lam * x)
u2 = sp.sin(sqrt_lam * (1 - x))
u1_at_xi = sp.sin(sqrt_lam * xi)
u2_at_xi = sp.sin(sqrt_lam * (1 - xi))

# Compute Wronskian at xi
du1_dxi = sp.diff(u1, x).subs(x, xi)
du2_dxi = sp.diff(u2, x).subs(x, xi)
W = u1_at_xi * du2_dxi - du1_dxi * u2_at_xi
W_simplified = sp.simplify(W)

print(f"Wronskian at xi: {W_simplified}")
# Output: Wronskian at xi: -0.841470984807897   (= -sin 1)
# With p = 1, the construction theorem gives
# G(x, xi) = u1(x) u2(xi) / W, and W = -sin(1) < 0:
# the Green's function is negative on (0, 1).

# Green's function
def greens_function(x_val, xi_val):
    W_val = -np.sin(sqrt_lam)   # p(xi) * W(xi) = -sin(1)
    if x_val <= xi_val:
        result = (np.sin(sqrt_lam * x_val) *
                 np.sin(sqrt_lam * (1 - xi_val)) / W_val)
    else:
        result = (np.sin(sqrt_lam * xi_val) *
                 np.sin(sqrt_lam * (1 - x_val)) / W_val)
    return result

# Verify the Green's function by checking the differential equation
x_sym = sp.symbols('x', real=True)
xi_val = 0.3
G = sp.Piecewise(
    (sp.sin(sqrt_lam * x_sym) *
     np.sin(sqrt_lam * (1 - xi_val)) / -np.sin(sqrt_lam), x_sym <= xi_val),
    (sp.sin(sqrt_lam * xi_val) *
     sp.sin(sqrt_lam * (1 - x_sym)) / -np.sin(sqrt_lam), x_sym > xi_val)
)

# Plot the Green's function for a fixed xi
x_vals = np.linspace(0, 1, 1000)
G_vals = np.array([greens_function(x_v, 0.3) for x_v in x_vals])

plt.figure(figsize=(10, 6))
plt.plot(x_vals, G_vals, 'b-', linewidth=2, label='$G(x, 0.3)$')
plt.axvline(x=0.3, color='r', linestyle='--', alpha=0.5, label='$x = \\xi = 0.3$')
plt.grid(True, alpha=0.3)
plt.xlabel('$x$', fontsize=12)
plt.ylabel('$G(x, \\xi)$', fontsize=12)
plt.title('Green\'s Function for Boundary Value Problem', fontsize=14)
plt.legend(fontsize=11)
plt.tight_layout()
plt.savefig('greens_function.png', dpi=150)
plt.close()

print("Green's function plotted successfully")
```

### Example: Beam Deflection Green's Function

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate

# Beam parameters
L = 10.0  # Length in meters
EI = 1.0  # Bending stiffness

def beam_greens_function(x, xi, L=10.0, EI=1.0):
    """
    Green's function for a simply supported beam:
      EI * y''''(x) = delta(x - xi),
      y(0) = y(L) = 0,  y''(0) = y''(L) = 0.
    Returns the deflection at position x due to a unit point
    load at xi. Satisfies G(0, xi) = G(L, xi) = 0.
    """
    if x <= xi:
        return x * (L - xi) * (2 * L * xi - xi**2 - x**2) / (6 * L * EI)
    else:
        return xi * (L - x) * (2 * L * x - x**2 - xi**2) / (6 * L * EI)

def beam_deflection_distributed_load(L, load_func):
    """
    Compute beam deflection for distributed load
    using Green's function approach
    """
    def integrand(xi, x):
        return beam_greens_function(x, xi, L) * load_func(xi)

    x_vals = np.linspace(0, L, 200)
    y_vals = []

    for x in x_vals:
        result, _ = integrate.quad(integrand, 0, L, args=(x,))
        y_vals.append(result)

    return x_vals, np.array(y_vals)

# Example: triangular distributed load
def triangular_load(x, L=10.0, max_load=1.0):
    """Triangular load with maximum at center"""
    return max_load * (1 - 4*(x - L/2)**2 / L**2) if 0 <= x <= L else 0

# Compute deflection
x_vals, y_vals = beam_deflection_distributed_load(L, triangular_load)

# Plotting
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

# Plot load
loads = np.array([triangular_load(x, L) for x in x_vals])
ax1.fill_between(x_vals, 0, loads, alpha=0.3, color='blue')
ax1.plot(x_vals, loads, 'b-', linewidth=2)
ax1.set_ylabel('Load (N/m)', fontsize=11)
ax1.set_title('Triangular Distributed Load', fontsize=12)
ax1.grid(True, alpha=0.3)

# Plot deflection
ax2.plot(x_vals, y_vals, 'r-', linewidth=2)
ax2.fill_between(x_vals, 0, y_vals, alpha=0.2, color='red')
ax2.set_xlabel('Position along beam (m)', fontsize=11)
ax2.set_ylabel('Deflection (m)', fontsize=11)
ax2.set_title('Beam Deflection via Green\'s Function', fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.invert_yaxis()

plt.tight_layout()
plt.savefig('beam_greens_function.png', dpi=150)
plt.close()

print("Beam deflection computed and plotted")
```

## Power Series Solutions to Differential Equations

### Python/SymPy Examples

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Define symbolic variables
x = sp.symbols('x')
n = sp.symbols('n', integer=True, positive=True)

# Method 1: Direct power series approach
# Assume y = sum of a_n * x^n from n=0 to infinity
# Then y' = sum of n * a_n * x^(n-1)
# and y'' = sum of n(n-1) * a_n * x^(n-2)

# For y'' + xy' + y = 0:
# sum of n(n-1)*a_n*x^(n-2) + sum of n*a_n*x^n + sum of a_n*x^n = 0

# Collecting the coefficient of x^n:
# (n+2)(n+1)*a_{n+2} + n*a_n + a_n = 0
# (n+2)(n+1)*a_{n+2} + (n+1)*a_n = 0
# a_{n+2} = -a_n / (n+2)

# This is a recurrence relation
def compute_power_series_coefficients(max_terms=10, a0=1, a1=0):
    """
    Compute coefficients for y'' + xy' + y = 0
    with y(0) = a0, y'(0) = a1
    """
    coeffs = [0] * max_terms
    coeffs[0] = a0
    if max_terms > 1:
        coeffs[1] = a1

    for k in range(max_terms - 2):
        coeffs[k + 2] = -coeffs[k] / (k + 2)

    return coeffs

# Compute coefficients
coeffs = compute_power_series_coefficients(max_terms=12, a0=1, a1=0)

# Display coefficients
print("Power Series Coefficients for y'' + xy' + y = 0:")
for i, c in enumerate(coeffs):
    print(f"a_{i} = {c:.6f}")

# Define the approximate solution
def power_series_solution(x_val, coeffs, max_terms=None):
    """Evaluate power series at x_val"""
    if max_terms is None:
        max_terms = len(coeffs)

    result = 0
    for k in range(min(max_terms, len(coeffs))):
        result += coeffs[k] * x_val**k
    return result

# Method 2: Using SymPy for verification
y = sp.Function('y')
ode = sp.Eq(y(x).diff(x, 2) + x*y(x).diff(x) + y(x), 0)

# Solve using power series
try:
    series_solution = sp.dsolve(ode, y(x), hint='series', n=5)
    print("\nSymPy Power Series Solution:")
    print(series_solution)
except:
    print("\nDirect series solution (computed recursively)")

# Plot the power series approximation
x_vals = np.linspace(-2, 2, 1000)
y_vals_4 = np.array([power_series_solution(x_v, coeffs, max_terms=4)
                      for x_v in x_vals])
y_vals_8 = np.array([power_series_solution(x_v, coeffs, max_terms=8)
                      for x_v in x_vals])
y_vals_12 = np.array([power_series_solution(x_v, coeffs, max_terms=12)
                       for x_v in x_vals])

plt.figure(figsize=(11, 6))
plt.plot(x_vals, y_vals_4, 'b-', linewidth=1.5, label='4 terms', alpha=0.7)
plt.plot(x_vals, y_vals_8, 'g-', linewidth=1.5, label='8 terms', alpha=0.7)
plt.plot(x_vals, y_vals_12, 'r-', linewidth=2, label='12 terms')
# For these initial conditions the exact solution is y = exp(-x^2/2)
plt.plot(x_vals, np.exp(-x_vals**2 / 2), 'k--', linewidth=1.5,
         label=r'Exact: $e^{-x^2/2}$')
plt.grid(True, alpha=0.3)
plt.xlabel('x', fontsize=12)
plt.ylabel('y(x)', fontsize=12)
plt.title('Power Series Solutions to $y\'\' + xy\' + y = 0$', fontsize=13)
plt.legend(fontsize=11)
plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
plt.ylim([-1.5, 1.5])
plt.tight_layout()
plt.savefig('power_series_solution.png', dpi=150)
plt.close()

print("\nPower series solution plotted")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import jn, yn, jvp, yvp

# Bessel's equation: x^2*y'' + x*y' + (x^2 - nu^2)*y = 0
# Solutions: J_nu(x) and Y_nu(x)

# Example 1: Vibrations of a circular drum head
# The displacement u(r,theta,t) satisfies wave equation in cylindrical coords
# This leads to Bessel's equation of order n

def drum_vibration(nu=0):
    """
    Compute vibration modes of a circular drum
    Boundary condition: u(a) = 0 (fixed edge)
    """
    x = np.linspace(0.01, 10, 1000)

    J_nu = jn(nu, x)
    Y_nu = yn(nu, x)

    return x, J_nu, Y_nu

# Compute Bessel functions
x_vals, J0, Y0 = drum_vibration(nu=0)
x_vals, J1, Y1 = drum_vibration(nu=1)
x_vals, J2, Y2 = drum_vibration(nu=2)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot Bessel functions of first kind
ax = axes[0]
ax.plot(x_vals, J0, 'b-', linewidth=2, label='$J_0(x)$')
ax.plot(x_vals, J1, 'g-', linewidth=2, label='$J_1(x)$')
ax.plot(x_vals, J2, 'r-', linewidth=2, label='$J_2(x)$')
ax.axhline(y=0, color='k', linestyle='-', alpha=0.2)
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('$J_n(x)$', fontsize=12)
ax.set_title('Bessel Functions of First Kind', fontsize=13)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
ax.set_xlim([0, 10])

# Plot Bessel functions of second kind
ax = axes[1]
ax.plot(x_vals, Y0, 'b-', linewidth=2, label='$Y_0(x)$')
ax.plot(x_vals, Y1, 'g-', linewidth=2, label='$Y_1(x)$')
ax.plot(x_vals, Y2, 'r-', linewidth=2, label='$Y_2(x)$')
ax.axhline(y=0, color='k', linestyle='-', alpha=0.2)
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('$Y_n(x)$', fontsize=12)
ax.set_title('Bessel Functions of Second Kind', fontsize=13)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
ax.set_xlim([0, 10])

plt.tight_layout()
plt.savefig('bessel_functions.png', dpi=150)
plt.close()

# Find zeros of J_n (drum vibration modes). SciPy provides
# scipy.special.jn_zeros, which returns the first k positive zeros
# directly -- preferable to writing a fragile root-finder by hand.
from scipy.special import jn_zeros

zeros_J0 = jn_zeros(0, 5)
zeros_J1 = jn_zeros(1, 5)

print("First 5 zeros of J0(x):", zeros_J0)
print("First 5 zeros of J1(x):", zeros_J1)

# Natural frequencies of drum with a = 1
c = 1  # wave speed
a = 1  # drum radius
freq_J0 = zeros_J0 * c / a

print("\nNatural frequencies (first mode, nu=0):")
for i, f in enumerate(freq_J0[:3]):
    print(f"Mode {i+1}: f = {f:.4f} (normalized)")

print("\nBessel function analysis complete")
```

## Nonhomogeneous Ordinary Differential Equations

### Method of Undetermined Coefficients

```python
import sympy as sp

# Define symbolic variables
x = sp.Symbol('x')
y = sp.Function('y')(x)

# Define the ODE: y'' - 3y' + 2y = 4e^(2x)
ode = sp.Eq(y.diff(x, 2) - 3*y.diff(x) + 2*y, 4*sp.exp(2*x))

# Solve the ODE
solution = sp.dsolve(ode, y)
print("General solution:")
print(solution)

# We can simplify/rewrite it to match our manual calculation
C1, C2 = sp.symbols('C1 C2')  # Integration constants
manual_form = C1*sp.exp(x) + C2*sp.exp(2*x) + 4*x*sp.exp(2*x)
print("\nSolution in our manual form:")
print(sp.Eq(y, manual_form))
```

```python
import sympy as sp

# Define symbolic variables
x = sp.Symbol('x')
y = sp.Function('y')(x)

# Define the ODE: y'' + 4y = 3sin(2x) + 2x^2
ode = sp.Eq(y.diff(x, 2) + 4*y, 3*sp.sin(2*x) + 2*x**2)

# Solve the ODE
solution = sp.dsolve(ode, y)
print("General solution from SymPy:")
print(solution)

# Simplify and verify our manual calculation
simplified = solution.rhs.expand().simplify()
print("\nSimplified form:")
print(simplified)
```

```python
import sympy as sp

# Define symbolic variables
x = sp.Symbol('x', real=True)
y = sp.Function('y')(x)

# Define the ODE: y'' + y = tan(x)
ode = sp.Eq(y.diff(x, 2) + y, sp.tan(x))

# Solve the ODE
solution = sp.dsolve(ode, y)
print("General solution:")
print(solution)

# Simplify and compare with our manual solution
simplified = solution.rhs.simplify()
print("\nSimplified form:")
print(simplified)

# Let's verify by substituting back into the original equation
y_sol = simplified
y_diff1 = sp.diff(y_sol, x)
y_diff2 = sp.diff(y_diff1, x)
original_eq = y_diff2 + y_sol - sp.tan(x)

print("\nVerification (should be approximately zero):")
print(original_eq.simplify())
```

### Python Implementation for Nonhomogeneous ODEs

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Example: y'' - 3y' + 2y = 4e^(2x) with y(0) = 1, y'(0) = 0
def nonhomogeneous_system(x, y):
    # Convert to first-order system:
    # y[0] = y, y[1] = y'
    # y'[0] = y[1], y'[1] = 3y[1] - 2y[0] + 4e^(2x)
    return [y[1], 3*y[1] - 2*y[0] + 4*np.exp(2*x)]

# Initial conditions: y(0) = 1, y'(0) = 0
y0 = [1, 0]

# Solve the ODE
x_span = [0, 2]
solution = solve_ivp(nonhomogeneous_system, x_span, y0, 
                    method='RK45', dense_output=True)

# Create a finer grid for plotting
x_plot = np.linspace(0, 2, 200)
y_plot = solution.sol(x_plot)

# Analytical solution for comparison
# y = c1*e^x + c2*e^(2x) + 4x*e^(2x); y(0)=1 and y'(0)=0
# give c1 + c2 = 1 and c1 + 2*c2 + 4 = 0, so c1 = 6, c2 = -5
def analytical_solution(x):
    c1 = 6
    c2 = -5
    return c1*np.exp(x) + c2*np.exp(2*x) + 4*x*np.exp(2*x)

y_analytical = analytical_solution(x_plot)

# Plot the numerical and analytical solutions
plt.figure(figsize=(10, 6))
plt.plot(x_plot, y_plot[0], 'b-', label='Numerical')
plt.plot(x_plot, y_analytical, 'r--', label='Analytical')
plt.xlabel('x')
plt.ylabel('y(x)')
plt.title('Solution to y\'\' - 3y\' + 2y = 4e^(2x)')
plt.grid(True)
plt.legend()
plt.show()
```

## Forced Oscillations

### Harmonic Forcing and Steady-State Response

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# System parameters
m = 1.0      # mass (kg)
k = 16.0     # spring constant (N/m)
c = 2.0      # damping coefficient (N*s/m)
F0 = 5.0     # forcing amplitude (N)
omega = 3.0  # forcing frequency (rad/s)

# Natural frequency and damping ratio
omega0 = np.sqrt(k/m)
zeta = c/(2*m*omega0)

# Calculate steady-state amplitude and phase
C = (F0/m)/np.sqrt((omega0**2 - omega**2)**2 + (2*zeta*omega0*omega)**2)
phi = np.arctan2(2*zeta*omega0*omega, omega0**2 - omega**2)

print(f"Natural frequency: {omega0} rad/s")
print(f"Damping ratio: {zeta}")
print(f"Steady-state amplitude: {C:.4f} m")
print(f"Phase lag: {phi:.4f} rad ({phi*180/np.pi:.2f} degrees)")

# System of first-order ODEs
def forced_oscillator(t, y):
    # y[0] = x, y[1] = x'
    return [
        y[1], 
        (F0*np.cos(omega*t) - c*y[1] - k*y[0])/m
    ]

# Initial conditions: x(0) = 0, x'(0) = 0
y0 = [0, 0]

# Time span: simulate for 10 seconds
t_span = [0, 10]

# Solve the ODE system
solution = solve_ivp(forced_oscillator, t_span, y0, 
                    method='RK45', t_eval=np.linspace(0, 10, 1000))

t = solution.t
x = solution.y[0]

# Steady-state analytical solution (cosine forcing -> cosine response)
t_analytical = np.linspace(0, 10, 1000)
x_analytical = C * np.cos(omega * t_analytical - phi)

# Plot the results
plt.figure(figsize=(12, 8))

# Transient and steady-state numerical solution
plt.subplot(2, 1, 1)
plt.plot(t, x, 'b-', label='Numerical (Full Response)')
plt.plot(t_analytical, x_analytical, 'r--', label='Analytical (Steady-State)')
plt.plot(t, F0/k * np.cos(omega*t), 'g:', label='Forcing Function (Scaled)')
plt.xlabel('Time (s)')
plt.ylabel('Position (m)')
plt.title('Forced Oscillations: Full Response vs. Steady-State')
plt.grid(True)
plt.legend()

# Zoom in on steady-state portion
plt.subplot(2, 1, 2)
t_steady = np.linspace(5, 10, 500)  # Last 5 seconds
x_steady = C * np.cos(omega * t_steady - phi)
plt.plot(t[t > 5], x[t > 5], 'b-', label='Numerical (Steady-State)')
plt.plot(t_steady, x_steady, 'r--', label='Analytical')
plt.xlabel('Time (s)')
plt.ylabel('Position (m)')
plt.title('Steady-State Response (Zoomed)')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

### Resonance Phenomenon

```python
import numpy as np
import matplotlib.pyplot as plt

# System parameters
m = 1.0      # mass (kg)
k = 16.0     # spring constant (N/m)
c = 2.0      # damping coefficient (N*s/m)
F0 = 5.0     # forcing amplitude (N)

# Natural frequency and damping ratio
omega0 = np.sqrt(k/m)
zeta = c/(2*m*omega0)

# Function to calculate amplitude
def amplitude(omega):
    return (F0/m)/np.sqrt((omega0**2 - omega**2)**2 + (2*zeta*omega0*omega)**2)

# Function to calculate phase
def phase(omega):
    return np.arctan2(2*zeta*omega0*omega, omega0**2 - omega**2)

# Create frequency range (amplitude and phase are vectorized)
omega_range = np.linspace(0.1, 8, 500)
amplitude_values = amplitude(omega_range)
phase_values = phase(omega_range) * 180/np.pi  # Convert to degrees

# Plot amplitude response
plt.figure(figsize=(12, 10))

plt.subplot(2, 1, 1)
plt.plot(omega_range, amplitude_values, 'b-', linewidth=2)
plt.axvline(x=omega0, color='r', linestyle='--', label=f'$\\omega_0 = {omega0}$')
plt.grid(True)
plt.xlabel('Forcing Frequency $\\omega$ (rad/s)')
plt.ylabel('Amplitude (m)')
plt.title('Amplitude Response (Resonance Curve)')
plt.legend()

# Add markers for specific frequencies
specific_omegas = [1, 2, 3, 4, 5, 6]
for w in specific_omegas:
    plt.plot(w, amplitude(w), 'ro')
    plt.text(w, amplitude(w)*1.1, f'$\\omega={w}$')

# Plot phase response
plt.subplot(2, 1, 2)
plt.plot(omega_range, phase_values, 'g-', linewidth=2)
plt.axvline(x=omega0, color='r', linestyle='--', label=f'$\\omega_0 = {omega0}$')
plt.axhline(y=90, color='k', linestyle=':', label='90')
plt.grid(True)
plt.xlabel('Forcing Frequency $\\omega$ (rad/s)')
plt.ylabel('Phase Angle (degrees)')
plt.title('Phase Response')
plt.ylim(0, 180)
plt.legend()

plt.tight_layout()
plt.show()
```

### Advanced Analysis: Frequency Response Function

```python
import sympy as sp

# Define symbolic variables
s, m, c, k, omega = sp.symbols('s m c k omega', real=True)
F0 = sp.symbols('F0', real=True)

# Transfer function in Laplace domain
H_s = 1/(m*s**2 + c*s + k)

# Substitute s = i*omega for frequency response
H_iw = H_s.subs(s, sp.I*omega)

# Simplify to standard form
H_iw = sp.simplify(H_iw)

# Convert to magnitude and phase
magnitude = sp.sqrt(H_iw * sp.conjugate(H_iw))
phase = sp.atan2(sp.im(H_iw), sp.re(H_iw))

# Print results
print("Frequency Response Function H(iw):")
print(H_iw)
print("\nMagnitude |H(iw)|:")
print(magnitude.simplify())
print("\nPhase H(iw):")
print(phase.simplify())
```

## Phase Diagrams

### Simple Harmonic Oscillator

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# System parameters
omega = 2.0  # natural frequency

# System of ODEs for simple harmonic oscillator
def harmonic_oscillator(t, y):
    # y[0] = x, y[1] = x'
    return [y[1], -omega**2 * y[0]]

# Function to plot trajectories for various initial conditions
def plot_trajectories(ax, initial_conditions):
    for ic in initial_conditions:
        # Solve ODE system
        sol = solve_ivp(harmonic_oscillator, [0, 10], ic, 
                       method='RK45', t_eval=np.linspace(0, 10, 1000))
        
        # Plot trajectory
        ax.plot(sol.y[0], sol.y[1], '-', linewidth=1.5)
        
        # Mark initial condition
        ax.plot(ic[0], ic[1], 'o', markersize=4)

# Draw vector field
def draw_vector_field(ax, x_range, y_range, density=20):
    x = np.linspace(x_range[0], x_range[1], density)
    y = np.linspace(y_range[0], y_range[1], density)
    X, Y = np.meshgrid(x, y)
    
    # Calculate derivatives at each point
    U = Y
    V = -omega**2 * X
    
    # Normalize vectors for clearer visualization
    magnitude = np.sqrt(U**2 + V**2)
    U = U / magnitude
    V = V / magnitude
    
    ax.quiver(X, Y, U, V, alpha=0.3)

# Create figure
fig, ax = plt.subplots(figsize=(10, 10))

# Phase space ranges
x_range = [-3, 3]
y_range = [-6, 6]

# Initial conditions: different amplitudes and phases
initial_conditions = [
    [2, 0],    # High energy, starting at positive displacement
    [0, 4],    # High energy, starting at zero displacement
    [1, 0],    # Medium energy, starting at positive displacement
    [0, 2],    # Medium energy, starting at zero displacement
    [0.5, 0],  # Low energy, starting at positive displacement
    [-1, -2],  # Mixed initial conditions
    [-2, 1]    # Mixed initial conditions
]

# Plot phase trajectories
plot_trajectories(ax, initial_conditions)

# Draw vector field
draw_vector_field(ax, x_range, y_range)

# Add theoretical ellipses
for C in [1, 2, 3]:
    theta = np.linspace(0, 2*np.pi, 100)
    x = C * np.cos(theta)
    y = -C * omega * np.sin(theta)
    ax.plot(x, y, 'r--', alpha=0.5)

# Add energy labels: E = 0.5*xdot^2 + 0.5*omega^2*x^2 = 2*C^2
# per unit mass, for an orbit of amplitude C with omega = 2
for C in [1, 2, 3]:
    ax.text(C+0.1, 0, f'E={2*C**2:.0f}', color='red')

# Configure plot
ax.set_xlim(x_range)
ax.set_ylim(y_range)
ax.set_xlabel('Position $x$')
ax.set_ylabel(r'Velocity $\dot{x}$')
ax.set_title('Phase Diagram for Simple Harmonic Oscillator')
ax.grid(True)
ax.set_aspect('equal')
ax.axhline(y=0, color='k', linestyle='-', alpha=0.3)
ax.axvline(x=0, color='k', linestyle='-', alpha=0.3)

plt.tight_layout()
plt.show()
```

### Damped Harmonic Oscillator

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Natural frequency
omega0 = 4.0

# Function to create damped oscillator system
def damped_oscillator(zeta):
    def system(t, y):
        # y[0] = x, y[1] = x'
        return [
            y[1],
            -omega0**2 * y[0] - 2*zeta*omega0 * y[1]
        ]
    return system

# Function to plot trajectories for a specific damping ratio
def plot_phase_portrait(zeta, ax, title):
    # Create system function with this damping ratio
    system = damped_oscillator(zeta)
    
    # Initial conditions - all starting with the same displacement but different velocities
    initial_conditions = [
        [2, 0],    # Zero initial velocity
        [2, 2],    # Small positive initial velocity
        [2, -2],   # Small negative initial velocity
        [2, 5],    # Large positive initial velocity
        [2, -5],   # Large negative initial velocity
        [-2, 0],   # Negative displacement, zero velocity
        [-2, 3]    # Negative displacement, positive velocity
    ]
    
    # Plot each trajectory
    for ic in initial_conditions:
        sol = solve_ivp(system, [0, 5], ic, method='RK45', t_eval=np.linspace(0, 5, 500))
        ax.plot(sol.y[0], sol.y[1], '-', linewidth=1.5)
        ax.plot(ic[0], ic[1], 'o', markersize=4)
    
    # Draw vector field
    x = np.linspace(-3, 3, 20)
    y = np.linspace(-6, 6, 20)
    X, Y = np.meshgrid(x, y)
    U = Y
    V = -omega0**2 * X - 2*zeta*omega0 * Y
    
    # Normalize vectors
    magnitude = np.sqrt(U**2 + V**2)
    U = U / magnitude
    V = V / magnitude
    
    ax.quiver(X, Y, U, V, alpha=0.3)
    
    # Configure plot
    ax.set_xlim(-3, 3)
    ax.set_ylim(-6, 6)
    ax.set_xlabel('Position $x$')
    ax.set_ylabel(r'Velocity $\dot{x}$')
    ax.set_title(title)
    ax.grid(True)
    ax.axhline(y=0, color='k', linestyle='-', alpha=0.3)
    ax.axvline(x=0, color='k', linestyle='-', alpha=0.3)

# Create figure with three subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Plot the three cases
plot_phase_portrait(0.2, axes[0], 'Underdamped ($\\zeta = 0.2$)')
plot_phase_portrait(1.0, axes[1], 'Critically Damped ($\\zeta = 1.0$)')
plot_phase_portrait(2.0, axes[2], 'Overdamped ($\\zeta = 2.0$)')

plt.tight_layout()
plt.show()
```

### Nonlinear Systems and Limit Cycles

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Van der Pol oscillator
def van_der_pol(mu):
    def system(t, y):
        # y[0] = x, y[1] = x'
        return [
            y[1],
            mu * (1 - y[0]**2) * y[1] - y[0]
        ]
    return system

# Create figure
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

# Plot for two different mu values
mu_values = [1.0, 5.0]

for i, mu in enumerate(mu_values):
    ax = axes[i]
    system = van_der_pol(mu)
    
    # Various initial conditions
    initial_conditions = [
        [0.5, 0],    # Inside the limit cycle
        [3.0, 0],    # Outside the limit cycle
        [0, 0.5],    # Different phase
        [0, 2.0],    # Different phase
        [0, -2.0],   # Different phase
        [-2.0, 1.0]  # Different quadrant
    ]
    
    # Time to simulate (longer for higher mu)
    t_max = 30 if mu == 1.0 else 60
    
    # Plot each trajectory
    for ic in initial_conditions:
        sol = solve_ivp(system, [0, t_max], ic, method='RK45', 
                       t_eval=np.linspace(0, t_max, 2000))
        ax.plot(sol.y[0], sol.y[1], '-', linewidth=1.0)
        ax.plot(ic[0], ic[1], 'o', markersize=4)
    
    # Draw vector field
    x = np.linspace(-4, 4, 20)
    y = np.linspace(-6, 6, 20)
    X, Y = np.meshgrid(x, y)
    U = Y
    V = mu * (1 - X**2) * Y - X
    
    # Normalize vectors
    magnitude = np.sqrt(U**2 + V**2)
    U = U / magnitude
    V = V / magnitude
    
    ax.quiver(X, Y, U, V, alpha=0.2)
    
    # Configure plot
    ax.set_xlim(-4, 4)
    ax.set_ylim(-6, 6)
    ax.set_xlabel('Position $x$')
    ax.set_ylabel(r'Velocity $\dot{x}$')
    ax.set_title(f'Van der Pol Oscillator ($\\mu = {mu}$)')
    ax.grid(True)
    ax.axhline(y=0, color='k', linestyle='-', alpha=0.3)
    ax.axvline(x=0, color='k', linestyle='-', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Numerical Methods for ODEs

### Euler's Method

```python
import numpy as np
import matplotlib.pyplot as plt

# Define our system of ODEs
def f(x, y):
    # y[0] = y, y[1] = y'
    return np.array([y[1], -y[0]])

# Euler method implementation
def euler_method(f, x0, y0, x_end, h):
    # Calculate number of steps
    n_steps = int((x_end - x0) / h)
    
    # Initialize arrays
    x = np.zeros(n_steps + 1)
    y = np.zeros((len(y0), n_steps + 1))
    
    # Set initial conditions
    x[0] = x0
    y[:, 0] = y0
    
    # Perform Euler integration
    for i in range(n_steps):
        y[:, i+1] = y[:, i] + h * f(x[i], y[:, i])
        x[i+1] = x[i] + h
    
    return x, y

# Initial conditions
x0 = 0
y0 = np.array([1.0, 0.0])  # y(0) = 1, y'(0) = 0
x_end = 2*np.pi
h = 0.1

# Apply Euler method
x_euler, y_euler = euler_method(f, x0, y0, x_end, h)

# Exact solution for comparison
x_exact = np.linspace(0, 2*np.pi, 1000)
y_exact = np.cos(x_exact)

# Plot results
plt.figure(figsize=(12, 6))
plt.plot(x_exact, y_exact, 'b-', label=r'Exact: $y = \cos(x)$')
plt.plot(x_euler, y_euler[0], 'ro', label=f'Euler: $h={h}$')
plt.xlabel('x')
plt.ylabel('y')
plt.title("Euler's Method vs. Exact Solution")
plt.grid(True)
plt.legend()

# Calculate and print error
final_error = abs(y_euler[0, -1] - np.cos(x_euler[-1]))
print(f"Final error at x = 2*pi: {final_error:.6f}")

plt.show()
```

### Runge-Kutta Methods

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# System parameters
damping = 0.1

# System of ODEs: y'' + 0.1y' + y = 0
def system(x, y):
    # y[0] = y, y[1] = y'
    return [y[1], -y[0] - damping * y[1]]

# Initial conditions
y0 = [1.0, 0.0]  # y(0) = 1, y'(0) = 0

# Euler method implementation
def euler_method(f, x0, y0, x_end, h):
    n_steps = int((x_end - x0) / h)
    x = np.zeros(n_steps + 1)
    y = np.zeros((len(y0), n_steps + 1))
    
    x[0] = x0
    y[:, 0] = y0
    
    for i in range(n_steps):
        y[:, i+1] = y[:, i] + h * np.array(f(x[i], y[:, i]))
        x[i+1] = x[i] + h
    
    return x, y

# RK4 method implementation
def rk4_method(f, x0, y0, x_end, h):
    n_steps = int((x_end - x0) / h)
    x = np.zeros(n_steps + 1)
    y = np.zeros((len(y0), n_steps + 1))
    
    x[0] = x0
    y[:, 0] = y0
    
    for i in range(n_steps):
        xi = x[i]
        yi = y[:, i]
        
        k1 = np.array(f(xi, yi))
        k2 = np.array(f(xi + h/2, yi + h*k1/2))
        k3 = np.array(f(xi + h/2, yi + h*k2/2))
        k4 = np.array(f(xi + h, yi + h*k3))
        
        y[:, i+1] = yi + (h/6) * (k1 + 2*k2 + 2*k3 + k4)
        x[i+1] = xi + h
    
    return x, y

# Integration parameters
x0 = 0
x_end = 20
h_euler = 0.05  # Smaller step for Euler for better accuracy
h_rk4 = 0.2     # RK4 can use larger steps

# Apply both methods
x_euler, y_euler = euler_method(system, x0, y0, x_end, h_euler)
x_rk4, y_rk4 = rk4_method(system, x0, y0, x_end, h_rk4)

# Use SciPy for reference solution (very accurate)
sol_ref = solve_ivp(system, [x0, x_end], y0, method='RK45', rtol=1e-12,
                   atol=1e-12, t_eval=np.linspace(x0, x_end, 1000))
x_ref = sol_ref.t
y_ref = sol_ref.y[0]

# Plot results
plt.figure(figsize=(12, 8))

# Position vs time
plt.subplot(2, 1, 1)
plt.plot(x_ref, y_ref, 'k-', label='Reference (SciPy)', linewidth=2)
plt.plot(x_euler, y_euler[0], 'b-', label=f'Euler (h={h_euler})', alpha=0.7)
plt.plot(x_rk4, y_rk4[0], 'r--', label=f'RK4 (h={h_rk4})', linewidth=2)
plt.xlabel('Time')
plt.ylabel('Position y(t)')
plt.title('Comparison of Numerical Methods for Damped Oscillator')
plt.grid(True)
plt.legend()

# Phase portrait
plt.subplot(2, 1, 2)
plt.plot(y_ref, sol_ref.y[1], 'k-', label='Reference (SciPy)', linewidth=2)
plt.plot(y_euler[0], y_euler[1], 'b-', label=f'Euler (h={h_euler})', alpha=0.7)
plt.plot(y_rk4[0], y_rk4[1], 'r--', label=f'RK4 (h={h_rk4})', linewidth=2)
plt.xlabel('Position y')
plt.ylabel('Velocity y\'')
plt.title('Phase Portrait')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()

# Calculate and print errors at final time
euler_error = abs(y_euler[0, -1] - y_ref[-1])
rk4_error = abs(y_rk4[0, -1] - y_ref[-1])
print(f"Euler final error: {euler_error:.8f}")
print(f"RK4 final error: {rk4_error:.8f}")
print(f"Error ratio: {euler_error/rk4_error:.1f}x")
```

### Stability and Error Analysis

#### Error Analysis

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Test problem: y' = -y with y(0) = 1, exact solution: y(x) = e^(-x)
def test_problem(x, y):
    return [-y[0]]

# Function to calculate error for different step sizes
def calculate_error(method, h_values):
    errors = []
    
    for h in h_values:
        if method.__name__ == 'euler_method':
            x, y = method(test_problem, 0, [1.0], 1.0, h)
            final_value = y[0, -1]
        else:  # RK4
            x, y = method(test_problem, 0, [1.0], 1.0, h)
            final_value = y[0, -1]
        
        exact_value = np.exp(-1.0)
        error = abs(final_value - exact_value)
        errors.append(error)
    
    return errors

# Step sizes to test
h_values = np.array([0.1, 0.05, 0.025, 0.0125, 0.00625])

# Calculate errors
euler_errors = calculate_error(euler_method, h_values)
rk4_errors = calculate_error(rk4_method, h_values)

# Plot errors vs step size on log-log scale
plt.figure(figsize=(10, 8))
plt.loglog(h_values, euler_errors, 'bo-', label='Euler Method')
plt.loglog(h_values, rk4_errors, 'rs-', label='RK4 Method')

# Add reference lines showing theoretical error rates
plt.loglog(h_values, h_values, 'k--', label='O(h)')
plt.loglog(h_values, h_values**2, 'g--', label='O(h^2)')
plt.loglog(h_values, h_values**4, 'm--', label='O(h^4)')

plt.xlabel('Step Size (h)')
plt.ylabel('Global Error at x=1')
plt.title('Error Convergence for Numerical Methods')
plt.grid(True)
plt.legend()

# Calculate and display slopes
euler_slope = np.polyfit(np.log(h_values), np.log(euler_errors), 1)[0]
rk4_slope = np.polyfit(np.log(h_values), np.log(rk4_errors), 1)[0]
plt.text(0.03, 1e-6, f'Euler slope = {euler_slope:.2f}', fontsize=12)
plt.text(0.03, 1e-9, f'RK4 slope = {rk4_slope:.2f}', fontsize=12)

plt.show()
```

#### Stability Analysis

```python
import numpy as np
import matplotlib.pyplot as plt

# Create grid of points in the complex plane
real = np.linspace(-3, 1, 200)
imag = np.linspace(-2, 2, 200)
re, im = np.meshgrid(real, imag)
z = re + 1j * im

# Stability functions
def euler_stability(z):
    return 1 + z

def rk4_stability(z):
    return 1 + z + z**2/2 + z**3/6 + z**4/24

# Calculate stability for each point
euler_stable = np.abs(euler_stability(z)) <= 1
rk4_stable = np.abs(rk4_stability(z)) <= 1

# Plot stability regions
plt.figure(figsize=(10, 8))
plt.contourf(re, im, euler_stable, levels=[0, 0.5, 1], alpha=0.3, colors=['white', 'blue'])
plt.contourf(re, im, rk4_stable, levels=[0, 0.5, 1], alpha=0.3, colors=['white', 'red'])
plt.contour(re, im, euler_stable, levels=[0.5], colors='blue', linewidths=2)
plt.contour(re, im, rk4_stable, levels=[0.5], colors='red', linewidths=2)

# Add labels and features. Filled contours carry no legend entries,
# so build the legend from proxy patches.
from matplotlib.patches import Patch
plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
plt.grid(True, alpha=0.3)
plt.xlabel('Re($h\\lambda$)')
plt.ylabel('Im($h\\lambda$)')
plt.title('Stability Regions for Numerical Methods')
plt.legend(handles=[Patch(color='blue', alpha=0.3, label='Euler stable'),
                    Patch(color='red', alpha=0.3, label='RK4 stable')])

# Show imaginary axis more clearly
plt.plot([0, 0], [-2, 2], 'k--', alpha=0.5)

# Add text labels
plt.text(-1.5, 0.5, 'Euler Stable', color='blue', fontsize=12)
plt.text(-2.5, 1.5, 'RK4 Stable', color='red', fontsize=12)

plt.tight_layout()
plt.show()
```

### Adaptive Step Size Methods

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Van der Pol oscillator (a system with varying timescales)
def van_der_pol(t, y, mu=10.0):
    return [
        y[1],
        mu * (1 - y[0]**2) * y[1] - y[0]
    ]

# Initial conditions
y0 = [2.0, 0.0]
t_span = [0, 40]

# Solve with fixed step size using the classical RK4 implementation
# from Section 3.9.2. SciPy's solve_ivp does NOT accept method='RK4'
# -- valid methods are RK23, RK45, DOP853, Radau, BDF, LSODA.
n_points = 500
h_fixed = (t_span[1] - t_span[0]) / n_points
t_fixed_arr, y_fixed_arr = rk4_method(van_der_pol, t_span[0], y0,
                                      t_span[1], h_fixed)

# Solve with adaptive step size RK45 (Dormand-Prince)
sol_adaptive = solve_ivp(van_der_pol, t_span, y0, method='RK45',
                        rtol=1e-6, atol=1e-9)

# Plot the results
plt.figure(figsize=(15, 10))

# Solution comparison
plt.subplot(2, 1, 1)
plt.plot(t_fixed_arr, y_fixed_arr[0], 'b-', label='Fixed step (classical RK4)')
plt.plot(sol_adaptive.t, sol_adaptive.y[0], 'ro', markersize=2,
        label=f'Adaptive step ({len(sol_adaptive.t)} points)')
plt.xlabel('Time')
plt.ylabel('Position x(t)')
plt.title('Van der Pol Oscillator Solution')
plt.grid(True)
plt.legend()

# Step size visualization
plt.subplot(2, 1, 2)
t_steps = sol_adaptive.t
step_sizes = np.diff(t_steps)
plt.semilogy(t_steps[:-1], step_sizes, 'k-')
plt.xlabel('Time')
plt.ylabel('Step Size (log scale)')
plt.title('Adaptive Step Size Variation')
plt.grid(True)

# Add time series above to show regions of rapid change
plt.twinx()
plt.plot(sol_adaptive.t, sol_adaptive.y[0], 'r-', alpha=0.3)
plt.ylabel('Position (for reference)', color='r')

plt.tight_layout()
plt.show()

# Quantify accuracy against a tight-tolerance reference
ref = solve_ivp(van_der_pol, t_span, y0, method='DOP853',
                rtol=1e-12, atol=1e-12, dense_output=True)
err_fixed = np.max(np.abs(y_fixed_arr[0] - ref.sol(t_fixed_arr)[0]))
err_adapt = np.max(np.abs(sol_adaptive.y[0] - ref.sol(sol_adaptive.t)[0]))
print(f"Fixed-step RK4: {n_points} points, max error {err_fixed:.1e}")
print(f"Adaptive RK45:  {len(sol_adaptive.t)} points, "
      f"max error {err_adapt:.1e}")
print(f"Adaptive step range: {np.diff(sol_adaptive.t).min():.1e}"
      f" to {np.diff(sol_adaptive.t).max():.1e}")
# Output:
# Fixed-step RK4: 500 points, max error 3.3e+00
# Adaptive RK45:  692 points, max error 1.7e-04
# Adaptive step range: 5.0e-04 to 1.1e-01
```
